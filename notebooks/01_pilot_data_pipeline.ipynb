```python
import csv
import json
from pathlib import Path
from collections import Counter

DATASET_DIR = Path(
    "/kaggle/input/datasets/tunguz/"
    "xview2-challenge-dataset-train-and-test"
)

IMAGE_DIR = DATASET_DIR / "train" / "train" / "images"
LABEL_DIR = DATASET_DIR / "train" / "train" / "labels"

OUTPUT_DIR = Path("/kaggle/working")
MANIFEST_PATH = OUTPUT_DIR / "dataset_manifest.csv"

rows = []
damage_counts = Counter()
missing_images = 0
invalid_annotations = 0

for label_path in sorted(LABEL_DIR.glob("*_post_disaster.json")):
    image_name = label_path.name.replace(".json", ".png")
    image_path = IMAGE_DIR / image_name

    if not image_path.exists():
        missing_images += 1
        continue

    try:
        with label_path.open("r", encoding="utf-8") as file:
            annotation = json.load(file)
    except (OSError, json.JSONDecodeError):
        invalid_annotations += 1
        continue

    features = annotation.get("features", {})
    buildings = features.get("xy", [])

    if not isinstance(buildings, list):
        invalid_annotations += 1
        continue

    for index, building in enumerate(buildings):
        properties = building.get("properties", {})
        if properties.get("feature_type") != "building":
            continue

        damage = properties.get("subtype")
        if not damage:
            continue

        polygon = building.get("wkt")
        if not polygon:
            continue

        rows.append({
            "image_path": str(image_path),
            "annotation_path": str(label_path),
            "image_name": image_name,
            "building_id": properties.get("uid", f"{image_name}_{index}"),
            "damage_label": damage,
            "polygon_wkt": polygon,
            "disaster": annotation.get("metadata", {}).get("disaster", ""),
            "disaster_type": annotation.get("metadata", {}).get("disaster_type", ""),
        })
        damage_counts[damage] += 1

columns = [
    "image_path",
    "annotation_path",
    "image_name",
    "building_id",
    "damage_label",
    "polygon_wkt",
    "disaster",
    "disaster_type",
]

with MANIFEST_PATH.open("w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=columns)
    writer.writeheader()
    writer.writerows(rows)

print("=== GeoVision Manifest Report ===")
print("Manifest saved to:", MANIFEST_PATH)
print("Building records:", len(rows))
print("Missing images:", missing_images)
print("Invalid annotation files:", invalid_annotations)
print("\nDamage label counts:")

for label, count in sorted(damage_counts.items()):
    print(f"{label}: {count}")

print("\nFirst three records:")
for row in rows[:3]:
    print(row["image_name"], "|", row["damage_label"], "|", row["building_id"])
```

```python

import json
import random
from pathlib import Path

import matplotlib.pyplot as plt
from PIL import Image
from matplotlib.patches import Polygon
from shapely import wkt

DATASET_DIR = Path(
    "/kaggle/input/datasets/tunguz/"
    "xview2-challenge-dataset-train-and-test"
)

IMAGE_DIR = DATASET_DIR / "train" / "train" / "images"
LABEL_DIR = DATASET_DIR / "train" / "train" / "labels"

# Choose one post-disaster annotation file
label_files = sorted(LABEL_DIR.glob("*_post_disaster.json"))
label_path = random.choice(label_files)

image_name = label_path.name.replace(".json", ".png")
image_path = IMAGE_DIR / image_name

with label_path.open("r", encoding="utf-8") as file:
    annotation = json.load(file)

image = Image.open(image_path).convert("RGB")

fig, ax = plt.subplots(figsize=(10, 10))
ax.imshow(image)

colors = {
    "no-damage": "lime",
    "minor-damage": "yellow",
    "major-damage": "orange",
    "destroyed": "red",
    "un-classified": "cyan",
}

shown = set()
polygon_count = 0

for building in annotation.get("features", {}).get("xy", []):
    properties = building.get("properties", {})

    if properties.get("feature_type") != "building":
        continue

    damage = properties.get("subtype", "un-classified")
    geometry_text = building.get("wkt")

    if not geometry_text:
        continue

    geometry = wkt.loads(geometry_text)

    polygons = (
        [geometry]
        if geometry.geom_type == "Polygon"
        else list(geometry.geoms)
        if geometry.geom_type == "MultiPolygon"
        else []
    )

    for polygon in polygons:
        coordinates = list(polygon.exterior.coords)

        patch = Polygon(
            coordinates,
            closed=True,
            fill=False,
            edgecolor=colors.get(damage, "white"),
            linewidth=0.7,
        )
        ax.add_patch(patch)
        polygon_count += 1

    shown.add(damage)

ax.set_title(f"{image_name}\nBuilding annotations: {polygon_count}")
ax.axis("off")
plt.show()

print("Image:", image_name)
print("Damage labels found:", sorted(shown))
```

```python

import json
import random
from pathlib import Path
from collections import Counter

import pandas as pd
from PIL import Image
from shapely import wkt

MANIFEST_PATH = Path("/kaggle/working/geo_vision_splits/train_manifest.csv")
OUTPUT_DIR = Path("/kaggle/working/building_crops_pilot")
CSV_PATH = OUTPUT_DIR / "crop_manifest.csv"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(MANIFEST_PATH)

# Pilot only: choose up to 500 building records
random.seed(42)
sample_df = df.sample(
    n=min(500, len(df)),
    random_state=42
)

crop_rows = []
counts = Counter()

for _, row in sample_df.iterrows():
    try:
        image_path = Path(row["image_path"])
        label_path = Path(row["annotation_path"])

        with label_path.open("r", encoding="utf-8") as f:
            annotation = json.load(f)

        image = Image.open(image_path).convert("RGB")
        buildings = annotation.get("features", {}).get("xy", [])

        # Find the selected building by its unique ID
        target = None
        for building in buildings:
            properties = building.get("properties", {})
            if properties.get("uid") == row["building_id"]:
                target = building
                break

        if target is None or not target.get("wkt"):
            continue

        polygon = wkt.loads(target["wkt"])
        min_x, min_y, max_x, max_y = polygon.bounds

        # Expand the crop slightly around the building
        padding = 8
        left = max(0, int(min_x) - padding)
        top = max(0, int(min_y) - padding)
        right = min(image.width, int(max_x) + padding + 1)
        bottom = min(image.height, int(max_y) + padding + 1)

        # Ignore invalid or extremely small crops
        if right - left < 8 or bottom - top < 8:
            continue

        crop = image.crop((left, top, right, bottom))

        # Save each crop inside its damage-class folder
        label = row["damage_label"]
        class_dir = OUTPUT_DIR / label
        class_dir.mkdir(parents=True, exist_ok=True)

        filename = f"{row['building_id']}.png"
        crop_path = class_dir / filename
        crop.save(crop_path)

        crop_rows.append({
            "crop_path": str(crop_path),
            "damage_label": label,
            "image_name": row["image_name"],
            "building_id": row["building_id"],
            "source_image": str(image_path),
        })
        counts[label] += 1

    except Exception as error:
        print("Skipped:", row["building_id"], "|", str(error))

crop_df = pd.DataFrame(crop_rows)
crop_df.to_csv(CSV_PATH, index=False)

print("=== GeoVision Pilot Crop Report ===")
print("Crops created:", len(crop_df))
print("Crop manifest:", CSV_PATH)
print("\nCrops per class:")
print(crop_df["damage_label"].value_counts())
```

```python

import pandas as pd
from pathlib import Path
from PIL import Image

CSV_PATH = Path(
    "/kaggle/working/building_crops_pilot/crop_manifest.csv"
)
OUTPUT_DIR = Path("/kaggle/working/building_crops_128")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(CSV_PATH)

updated_rows = []
small_crop_count = 0
failed_count = 0

for _, row in df.iterrows():
    try:
        source_path = Path(row["crop_path"])

        with Image.open(source_path) as image:
            image = image.convert("RGB")
            original_width, original_height = image.size

            if min(original_width, original_height) < 16:
                small_crop_count += 1

            resized = image.resize(
                (128, 128),
                Image.Resampling.BILINEAR
            )

            class_dir = OUTPUT_DIR / row["damage_label"]
            class_dir.mkdir(parents=True, exist_ok=True)

            output_path = class_dir / source_path.name
            resized.save(output_path)

        updated_row = row.to_dict()
        updated_row["resized_path"] = str(output_path)
        updated_row["original_width"] = original_width
        updated_row["original_height"] = original_height
        updated_rows.append(updated_row)

    except Exception as error:
        failed_count += 1
        print("Failed:", row["crop_path"], "|", error)

resized_df = pd.DataFrame(updated_rows)
output_csv = OUTPUT_DIR / "crop_manifest_128.csv"
resized_df.to_csv(output_csv, index=False)

print("=== GeoVision Resize Report ===")
print("Original crops:", len(df))
print("Successfully resized:", len(resized_df))
print("Crops with a dimension below 16 pixels:", small_crop_count)
print("Failed crops:", failed_count)
print("Output manifest:", output_csv)
print("\nResized dimensions:")
print(
    resized_df["resized_path"].apply(
        lambda path: Image.open(path).size
    ).value_counts()
)
```

```python

import random
import numpy as np
import pandas as pd
from pathlib import Path
from PIL import Image

import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import classification_report, confusion_matrix

# Reproducibility
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

DEVICE = torch.device("cpu")
print("Training device:", DEVICE)

CSV_PATH = Path(
    "/kaggle/working/building_crops_128/crop_manifest_128.csv"
)

df = pd.read_csv(CSV_PATH)

# Keep the four target classes and exclude tiny original crops
df = df[
    df["damage_label"].isin([
        "no-damage",
        "minor-damage",
        "major-damage",
        "destroyed",
    ])
].copy()

df = df[
    (df["original_width"] >= 16) &
    (df["original_height"] >= 16)
].copy()

classes = [
    "no-damage",
    "minor-damage",
    "major-damage",
    "destroyed",
]
label_to_id = {name: i for i, name in enumerate(classes)}
df["label_id"] = df["damage_label"].map(label_to_id)

# Split by original satellite image
splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=SEED,
)

train_idx, val_idx = next(
    splitter.split(df, groups=df["image_name"])
)

train_df = df.iloc[train_idx].reset_index(drop=True)
val_df = df.iloc[val_idx].reset_index(drop=True)

print("Training crops:", len(train_df))
print("Validation crops:", len(val_df))
print(
    "Shared source images:",
    len(set(train_df["image_name"]) & set(val_df["image_name"]))
)

# Image preprocessing
transform = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.ToTensor(),
])

class BuildingCropDataset(Dataset):
    def __init__(self, frame):
        self.frame = frame
        self.transform = transform

    def __len__(self):
        return len(self.frame)

    def __getitem__(self, index):
        row = self.frame.iloc[index]

        with Image.open(row["resized_path"]) as image:
            image = image.convert("RGB")
            image = self.transform(image)

        return image, int(row["label_id"])

train_dataset = BuildingCropDataset(train_df)
val_dataset = BuildingCropDataset(val_df)

train_loader = DataLoader(
    train_dataset,
    batch_size=16,
    shuffle=True,
    num_workers=0,
)

val_loader = DataLoader(
    val_dataset,
    batch_size=16,
    shuffle=False,
    num_workers=0,
)

# Small CNN
class DamageCNN(nn.Module):
    def __init__(self, num_classes=4):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 16, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.AdaptiveAvgPool2d((1, 1)),
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(32, num_classes),
        )

    def forward(self, x):
        return self.classifier(self.features(x))

model = DamageCNN().to(DEVICE)

# Class weights reduce the effect of class imbalance
counts = train_df["label_id"].value_counts().reindex(
    range(len(classes)), fill_value=1
)
weights = len(train_df) / (len(classes) * counts)
class_weights = torch.tensor(
    weights.to_numpy(), dtype=torch.float32
).to(DEVICE)

criterion = nn.CrossEntropyLoss(weight=class_weights)
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

# Train for a few epochs
EPOCHS = 5

for epoch in range(EPOCHS):
    model.train()
    total_loss = 0.0

    for images, labels in train_loader:
        images = images.to(DEVICE)
        labels = labels.to(DEVICE)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        total_loss += loss.item() * images.size(0)

    average_loss = total_loss / len(train_dataset)
    print(
        f"Epoch {epoch + 1}/{EPOCHS} "
        f"- Training loss: {average_loss:.4f}"
    )

# Evaluate on the held-out validation crops
model.eval()
all_predictions = []
all_labels = []

with torch.no_grad():
    for images, labels in val_loader:
        outputs = model(images.to(DEVICE))
        predictions = outputs.argmax(dim=1).cpu().numpy()

        all_predictions.extend(predictions)
        all_labels.extend(labels.numpy())

print("\n=== GeoVision Pilot Evaluation ===")
print(classification_report(
    all_labels,
    all_predictions,
    labels=list(range(len(classes))),
    target_names=classes,
    zero_division=0,
))

print("Confusion matrix (true labels = rows):")
print(confusion_matrix(
    all_labels,
    all_predictions,
    labels=list(range(len(classes))),
))

# Save the pilot model
MODEL_PATH = Path("/kaggle/working/geovision_pilot_cnn.pth")
torch.save({
    "model_state_dict": model.state_dict(),
    "classes": classes,
    "input_size": 128,
}, MODEL_PATH)

print("\nModel saved to:", MODEL_PATH)
```

```python

import numpy as np
import torch
from torch import nn
from torch.utils.data import DataLoader, WeightedRandomSampler
from sklearn.metrics import f1_score, classification_report

torch.manual_seed(42)
np.random.seed(42)

# Give minority classes more opportunities during training
train_labels = train_df["label_id"].to_numpy()
class_counts = np.bincount(train_labels, minlength=len(classes))

sample_weights = 1.0 / class_counts[train_labels]

sampler = WeightedRandomSampler(
    weights=torch.as_tensor(sample_weights, dtype=torch.double),
    num_samples=len(train_labels),
    replacement=True,
)

balanced_train_loader = DataLoader(
    train_dataset,
    batch_size=16,
    sampler=sampler,
    num_workers=0,
)

# Reinitialize the model for a fair new experiment
model = DamageCNN().to(DEVICE)

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.0005,
    weight_decay=0.0001,
)

EPOCHS = 15
best_macro_f1 = -1.0
best_state = None

for epoch in range(EPOCHS):
    model.train()
    total_loss = 0.0

    for images, labels in balanced_train_loader:
        images = images.to(DEVICE)
        labels = labels.to(DEVICE)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        total_loss += loss.item() * images.size(0)

    # Evaluate on the same held-out validation set
    model.eval()
    true_labels = []
    predictions = []

    with torch.no_grad():
        for images, labels in val_loader:
            outputs = model(images.to(DEVICE))
            predictions.extend(outputs.argmax(dim=1).cpu().tolist())
            true_labels.extend(labels.tolist())

    macro_f1 = f1_score(
        true_labels,
        predictions,
        labels=list(range(len(classes))),
        average="macro",
        zero_division=0,
    )

    avg_loss = total_loss / len(train_dataset)
    print(
        f"Epoch {epoch + 1:02d}/{EPOCHS} | "
        f"Loss: {avg_loss:.4f} | "
        f"Validation macro F1: {macro_f1:.3f}"
    )

    if macro_f1 > best_macro_f1:
        best_macro_f1 = macro_f1
        best_state = {
            key: value.detach().cpu().clone()
            for key, value in model.state_dict().items()
        }

# Restore the best checkpoint from this experiment
if best_state is not None:
    model.load_state_dict(best_state)

model.eval()
final_true = []
final_pred = []

with torch.no_grad():
    for images, labels in val_loader:
        outputs = model(images.to(DEVICE))
        final_pred.extend(outputs.argmax(dim=1).cpu().tolist())
        final_true.extend(labels.tolist())

print("\n=== Balanced Training Results ===")
print("Best validation macro F1:", round(best_macro_f1, 4))
print(classification_report(
    final_true,
    final_pred,
    labels=list(range(len(classes))),
    target_names=classes,
    zero_division=0,
))

MODEL_PATH = "/kaggle/working/geovision_pilot_cnn_balanced.pth"

torch.save({
    "model_state_dict": model.state_dict(),
    "classes": classes,
    "input_size": 128,
    "best_validation_macro_f1": best_macro_f1,
}, MODEL_PATH)

print("Saved model:", MODEL_PATH)
```

```python

import json
import pandas as pd
from pathlib import Path
from collections import Counter
from PIL import Image
from shapely import wkt

BASE_DIR = Path("/kaggle/working/geovision_large_pilot")
OUTPUT_DIR = BASE_DIR / "crops"

manifests = {
    "train": BASE_DIR / "train_manifest.csv",
    "val": BASE_DIR / "val_manifest.csv",
}

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

all_counts = Counter()
all_rows = []
failed = 0
padding = 8

for split, manifest_path in manifests.items():
    df = pd.read_csv(manifest_path)
    split_rows = []

    print(f"Processing {split}: {len(df)} buildings")

    for _, row in df.iterrows():
        try:
            image_path = Path(row["image_path"])
            label_path = Path(row["annotation_path"])

            with label_path.open("r", encoding="utf-8") as f:
                annotation = json.load(f)

            target = None
            for building in annotation.get(
                "features", {}
            ).get("xy", []):
                properties = building.get("properties", {})
                if properties.get("uid") == row["building_id"]:
                    target = building
                    break

            if target is None or not target.get("wkt"):
                failed += 1
                continue

            polygon = wkt.loads(target["wkt"])
            min_x, min_y, max_x, max_y = polygon.bounds

            with Image.open(image_path) as source:
                image = source.convert("RGB")

            left = max(0, int(min_x) - padding)
            top = max(0, int(min_y) - padding)
            right = min(image.width, int(max_x) + padding + 1)
            bottom = min(image.height, int(max_y) + padding + 1)

            if right - left < 16 or bottom - top < 16:
                failed += 1
                continue

            crop = image.crop((left, top, right, bottom))

            label = row["damage_label"]
            class_dir = OUTPUT_DIR / split / label
            class_dir.mkdir(parents=True, exist_ok=True)

            filename = f"{row['building_id']}.png"
            crop_path = class_dir / filename
            crop.save(crop_path)

            record = {
                "crop_path": str(crop_path),
                "damage_label": label,
                "image_name": row["image_name"],
                "building_id": row["building_id"],
                "original_width": crop.width,
                "original_height": crop.height,
            }

            split_rows.append(record)
            all_rows.append({"split": split, **record})
            all_counts[f"{split}/{label}"] += 1

        except Exception as error:
            failed += 1
            print("Skipped:", row["building_id"], "|", str(error))

    pd.DataFrame(split_rows).to_csv(
        BASE_DIR / f"{split}_crops_manifest.csv",
        index=False,
    )

    print(f"Saved {len(split_rows)} {split} crops")

print("\n=== GeoVision Crop Extraction Report ===")
print("Total crops created:", len(all_rows))
print("Failed or skipped records:", failed)

print("\nCrops per split and class:")
for key, count in sorted(all_counts.items()):
    print(f"{key}: {count}")

print("\nOutput folder:", OUTPUT_DIR)
print("Training manifest:", BASE_DIR / "train_crops_manifest.csv")
print("Validation manifest:", BASE_DIR / "val_crops_manifest.csv")
```

```python

import pandas as pd
from pathlib import Path
from PIL import Image

BASE_DIR = Path("/kaggle/working/geovision_large_pilot")
OUTPUT_DIR = BASE_DIR / "crops_128"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

manifests = {
    "train": BASE_DIR / "train_crops_manifest.csv",
    "val": BASE_DIR / "val_crops_manifest.csv",
}

for split, manifest_path in manifests.items():
    df = pd.read_csv(manifest_path)
    resized_rows = []
    failures = 0

    print(f"Resizing {split}: {len(df)} crops")

    for _, row in df.iterrows():
        try:
            with Image.open(row["crop_path"]) as image:
                image = image.convert("RGB")
                original_width, original_height = image.size

                resized = image.resize(
                    (128, 128),
                    Image.Resampling.BILINEAR,
                )

                output_path = (
                    OUTPUT_DIR
                    / split
                    / row["damage_label"]
                    / Path(row["crop_path"]).name
                )
                output_path.parent.mkdir(
                    parents=True,
                    exist_ok=True,
                )
                resized.save(output_path)

            record = row.to_dict()
            record["resized_path"] = str(output_path)
            record["original_width"] = original_width
            record["original_height"] = original_height
            resized_rows.append(record)

        except Exception as error:
            failures += 1
            print("Failed:", row["crop_path"], "|", error)

    output_csv = BASE_DIR / f"{split}_crops_128_manifest.csv"
    pd.DataFrame(resized_rows).to_csv(output_csv, index=False)

    print(f"Successfully resized: {len(resized_rows)}")
    print(f"Failures: {failures}")
    print(f"Manifest: {output_csv}")

print("\nResize operation finished.")
```
