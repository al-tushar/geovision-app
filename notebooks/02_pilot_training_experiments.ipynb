```python
import csv
import json
from pathlib import Path
from collections import Counter

DATASET_DIR = Path(
    "/kaggle/input/datasets/tunguz/"
    "xview2-challenge-dataset-train-and-test"
)

IMAGE_DIR = DATASET_DIR / "train" / "train" / "images"
LABEL_DIR = DATASET_DIR / "train" / "train" / "labels"

OUTPUT_DIR = Path("/kaggle/working")
MANIFEST_PATH = OUTPUT_DIR / "dataset_manifest.csv"

rows = []
damage_counts = Counter()
missing_images = 0
invalid_annotations = 0

for label_path in sorted(LABEL_DIR.glob("*_post_disaster.json")):
    image_name = label_path.name.replace(".json", ".png")
    image_path = IMAGE_DIR / image_name

    if not image_path.exists():
        missing_images += 1
        continue

    try:
        with label_path.open("r", encoding="utf-8") as file:
            annotation = json.load(file)
    except (OSError, json.JSONDecodeError):
        invalid_annotations += 1
        continue

    features = annotation.get("features", {})
    buildings = features.get("xy", [])

    if not isinstance(buildings, list):
        invalid_annotations += 1
        continue

    for index, building in enumerate(buildings):
        properties = building.get("properties", {})
        if properties.get("feature_type") != "building":
            continue

        damage = properties.get("subtype")
        if not damage:
            continue

        polygon = building.get("wkt")
        if not polygon:
            continue

        rows.append({
            "image_path": str(image_path),
            "annotation_path": str(label_path),
            "image_name": image_name,
            "building_id": properties.get("uid", f"{image_name}_{index}"),
            "damage_label": damage,
            "polygon_wkt": polygon,
            "disaster": annotation.get("metadata", {}).get("disaster", ""),
            "disaster_type": annotation.get("metadata", {}).get("disaster_type", ""),
        })
        damage_counts[damage] += 1

columns = [
    "image_path",
    "annotation_path",
    "image_name",
    "building_id",
    "damage_label",
    "polygon_wkt",
    "disaster",
    "disaster_type",
]

with MANIFEST_PATH.open("w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=columns)
    writer.writeheader()
    writer.writerows(rows)

print("=== GeoVision Manifest Report ===")
print("Manifest saved to:", MANIFEST_PATH)
print("Building records:", len(rows))
print("Missing images:", missing_images)
print("Invalid annotation files:", invalid_annotations)
print("\nDamage label counts:")

for label, count in sorted(damage_counts.items()):
    print(f"{label}: {count}")

print("\nFirst three records:")
for row in rows[:3]:
    print(row["image_name"], "|", row["damage_label"], "|", row["building_id"])
```

```python

import json
import random
from pathlib import Path

import matplotlib.pyplot as plt
from PIL import Image
from matplotlib.patches import Polygon
from shapely import wkt

DATASET_DIR = Path(
    "/kaggle/input/datasets/tunguz/"
    "xview2-challenge-dataset-train-and-test"
)

IMAGE_DIR = DATASET_DIR / "train" / "train" / "images"
LABEL_DIR = DATASET_DIR / "train" / "train" / "labels"

# Choose one post-disaster annotation file
label_files = sorted(LABEL_DIR.glob("*_post_disaster.json"))
label_path = random.choice(label_files)

image_name = label_path.name.replace(".json", ".png")
image_path = IMAGE_DIR / image_name

with label_path.open("r", encoding="utf-8") as file:
    annotation = json.load(file)

image = Image.open(image_path).convert("RGB")

fig, ax = plt.subplots(figsize=(10, 10))
ax.imshow(image)

colors = {
    "no-damage": "lime",
    "minor-damage": "yellow",
    "major-damage": "orange",
    "destroyed": "red",
    "un-classified": "cyan",
}

shown = set()
polygon_count = 0

for building in annotation.get("features", {}).get("xy", []):
    properties = building.get("properties", {})

    if properties.get("feature_type") != "building":
        continue

    damage = properties.get("subtype", "un-classified")
    geometry_text = building.get("wkt")

    if not geometry_text:
        continue

    geometry = wkt.loads(geometry_text)

    polygons = (
        [geometry]
        if geometry.geom_type == "Polygon"
        else list(geometry.geoms)
        if geometry.geom_type == "MultiPolygon"
        else []
    )

    for polygon in polygons:
        coordinates = list(polygon.exterior.coords)

        patch = Polygon(
            coordinates,
            closed=True,
            fill=False,
            edgecolor=colors.get(damage, "white"),
            linewidth=0.7,
        )
        ax.add_patch(patch)
        polygon_count += 1

    shown.add(damage)

ax.set_title(f"{image_name}\nBuilding annotations: {polygon_count}")
ax.axis("off")
plt.show()

print("Image:", image_name)
print("Damage labels found:", sorted(shown))
```

```python

import json
import random
from pathlib import Path
from collections import Counter

import pandas as pd
from PIL import Image
from shapely import wkt

MANIFEST_PATH = Path("/kaggle/working/geo_vision_splits/train_manifest.csv")
OUTPUT_DIR = Path("/kaggle/working/building_crops_pilot")
CSV_PATH = OUTPUT_DIR / "crop_manifest.csv"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(MANIFEST_PATH)

# Pilot only: choose up to 500 building records
random.seed(42)
sample_df = df.sample(
    n=min(500, len(df)),
    random_state=42
)

crop_rows = []
counts = Counter()

for _, row in sample_df.iterrows():
    try:
        image_path = Path(row["image_path"])
        label_path = Path(row["annotation_path"])

        with label_path.open("r", encoding="utf-8") as f:
            annotation = json.load(f)

        image = Image.open(image_path).convert("RGB")
        buildings = annotation.get("features", {}).get("xy", [])

        # Find the selected building by its unique ID
        target = None
        for building in buildings:
            properties = building.get("properties", {})
            if properties.get("uid") == row["building_id"]:
                target = building
                break

        if target is None or not target.get("wkt"):
            continue

        polygon = wkt.loads(target["wkt"])
        min_x, min_y, max_x, max_y = polygon.bounds

        # Expand the crop slightly around the building
        padding = 8
        left = max(0, int(min_x) - padding)
        top = max(0, int(min_y) - padding)
        right = min(image.width, int(max_x) + padding + 1)
        bottom = min(image.height, int(max_y) + padding + 1)

        # Ignore invalid or extremely small crops
        if right - left < 8 or bottom - top < 8:
            continue

        crop = image.crop((left, top, right, bottom))

        # Save each crop inside its damage-class folder
        label = row["damage_label"]
        class_dir = OUTPUT_DIR / label
        class_dir.mkdir(parents=True, exist_ok=True)

        filename = f"{row['building_id']}.png"
        crop_path = class_dir / filename
        crop.save(crop_path)

        crop_rows.append({
            "crop_path": str(crop_path),
            "damage_label": label,
            "image_name": row["image_name"],
            "building_id": row["building_id"],
            "source_image": str(image_path),
        })
        counts[label] += 1

    except Exception as error:
        print("Skipped:", row["building_id"], "|", str(error))

crop_df = pd.DataFrame(crop_rows)
crop_df.to_csv(CSV_PATH, index=False)

print("=== GeoVision Pilot Crop Report ===")
print("Crops created:", len(crop_df))
print("Crop manifest:", CSV_PATH)
print("\nCrops per class:")
print(crop_df["damage_label"].value_counts())
```

```python

import pandas as pd
from pathlib import Path
from PIL import Image

CSV_PATH = Path(
    "/kaggle/working/building_crops_pilot/crop_manifest.csv"
)
OUTPUT_DIR = Path("/kaggle/working/building_crops_128")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(CSV_PATH)

updated_rows = []
small_crop_count = 0
failed_count = 0

for _, row in df.iterrows():
    try:
        source_path = Path(row["crop_path"])

        with Image.open(source_path) as image:
            image = image.convert("RGB")
            original_width, original_height = image.size

            if min(original_width, original_height) < 16:
                small_crop_count += 1

            resized = image.resize(
                (128, 128),
                Image.Resampling.BILINEAR
            )

            class_dir = OUTPUT_DIR / row["damage_label"]
            class_dir.mkdir(parents=True, exist_ok=True)

            output_path = class_dir / source_path.name
            resized.save(output_path)

        updated_row = row.to_dict()
        updated_row["resized_path"] = str(output_path)
        updated_row["original_width"] = original_width
        updated_row["original_height"] = original_height
        updated_rows.append(updated_row)

    except Exception as error:
        failed_count += 1
        print("Failed:", row["crop_path"], "|", error)

resized_df = pd.DataFrame(updated_rows)
output_csv = OUTPUT_DIR / "crop_manifest_128.csv"
resized_df.to_csv(output_csv, index=False)

print("=== GeoVision Resize Report ===")
print("Original crops:", len(df))
print("Successfully resized:", len(resized_df))
print("Crops with a dimension below 16 pixels:", small_crop_count)
print("Failed crops:", failed_count)
print("Output manifest:", output_csv)
print("\nResized dimensions:")
print(
    resized_df["resized_path"].apply(
        lambda path: Image.open(path).size
    ).value_counts()
)
```

```python

import random
import numpy as np
import pandas as pd
from pathlib import Path
from PIL import Image

import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import classification_report, confusion_matrix

# Reproducibility
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

DEVICE = torch.device("cpu")
print("Training device:", DEVICE)

CSV_PATH = Path(
    "/kaggle/working/building_crops_128/crop_manifest_128.csv"
)

df = pd.read_csv(CSV_PATH)

# Keep the four target classes and exclude tiny original crops
df = df[
    df["damage_label"].isin([
        "no-damage",
        "minor-damage",
        "major-damage",
        "destroyed",
    ])
].copy()

df = df[
    (df["original_width"] >= 16) &
    (df["original_height"] >= 16)
].copy()

classes = [
    "no-damage",
    "minor-damage",
    "major-damage",
    "destroyed",
]
label_to_id = {name: i for i, name in enumerate(classes)}
df["label_id"] = df["damage_label"].map(label_to_id)

# Split by original satellite image
splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=SEED,
)

train_idx, val_idx = next(
    splitter.split(df, groups=df["image_name"])
)

train_df = df.iloc[train_idx].reset_index(drop=True)
val_df = df.iloc[val_idx].reset_index(drop=True)

print("Training crops:", len(train_df))
print("Validation crops:", len(val_df))
print(
    "Shared source images:",
    len(set(train_df["image_name"]) & set(val_df["image_name"]))
)

# Image preprocessing
transform = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.ToTensor(),
])

class BuildingCropDataset(Dataset):
    def __init__(self, frame):
        self.frame = frame
        self.transform = transform

    def __len__(self):
        return len(self.frame)

    def __getitem__(self, index):
        row = self.frame.iloc[index]

        with Image.open(row["resized_path"]) as image:
            image = image.convert("RGB")
            image = self.transform(image)

        return image, int(row["label_id"])

train_dataset = BuildingCropDataset(train_df)
val_dataset = BuildingCropDataset(val_df)

train_loader = DataLoader(
    train_dataset,
    batch_size=16,
    shuffle=True,
    num_workers=0,
)

val_loader = DataLoader(
    val_dataset,
    batch_size=16,
    shuffle=False,
    num_workers=0,
)

# Small CNN
class DamageCNN(nn.Module):
    def __init__(self, num_classes=4):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 16, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.AdaptiveAvgPool2d((1, 1)),
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(32, num_classes),
        )

    def forward(self, x):
        return self.classifier(self.features(x))

model = DamageCNN().to(DEVICE)

# Class weights reduce the effect of class imbalance
counts = train_df["label_id"].value_counts().reindex(
    range(len(classes)), fill_value=1
)
weights = len(train_df) / (len(classes) * counts)
class_weights = torch.tensor(
    weights.to_numpy(), dtype=torch.float32
).to(DEVICE)

criterion = nn.CrossEntropyLoss(weight=class_weights)
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

# Train for a few epochs
EPOCHS = 5

for epoch in range(EPOCHS):
    model.train()
    total_loss = 0.0

    for images, labels in train_loader:
        images = images.to(DEVICE)
        labels = labels.to(DEVICE)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        total_loss += loss.item() * images.size(0)

    average_loss = total_loss / len(train_dataset)
    print(
        f"Epoch {epoch + 1}/{EPOCHS} "
        f"- Training loss: {average_loss:.4f}"
    )

# Evaluate on the held-out validation crops
model.eval()
all_predictions = []
all_labels = []

with torch.no_grad():
    for images, labels in val_loader:
        outputs = model(images.to(DEVICE))
        predictions = outputs.argmax(dim=1).cpu().numpy()

        all_predictions.extend(predictions)
        all_labels.extend(labels.numpy())

print("\n=== GeoVision Pilot Evaluation ===")
print(classification_report(
    all_labels,
    all_predictions,
    labels=list(range(len(classes))),
    target_names=classes,
    zero_division=0,
))

print("Confusion matrix (true labels = rows):")
print(confusion_matrix(
    all_labels,
    all_predictions,
    labels=list(range(len(classes))),
))

# Save the pilot model
MODEL_PATH = Path("/kaggle/working/geovision_pilot_cnn.pth")
torch.save({
    "model_state_dict": model.state_dict(),
    "classes": classes,
    "input_size": 128,
}, MODEL_PATH)

print("\nModel saved to:", MODEL_PATH)
```

```python

import numpy as np
import torch
from torch import nn
from torch.utils.data import DataLoader, WeightedRandomSampler
from sklearn.metrics import f1_score, classification_report

torch.manual_seed(42)
np.random.seed(42)

# Give minority classes more opportunities during training
train_labels = train_df["label_id"].to_numpy()
class_counts = np.bincount(train_labels, minlength=len(classes))

sample_weights = 1.0 / class_counts[train_labels]

sampler = WeightedRandomSampler(
    weights=torch.as_tensor(sample_weights, dtype=torch.double),
    num_samples=len(train_labels),
    replacement=True,
)

balanced_train_loader = DataLoader(
    train_dataset,
    batch_size=16,
    sampler=sampler,
    num_workers=0,
)

# Reinitialize the model for a fair new experiment
model = DamageCNN().to(DEVICE)

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.0005,
    weight_decay=0.0001,
)

EPOCHS = 15
best_macro_f1 = -1.0
best_state = None

for epoch in range(EPOCHS):
    model.train()
    total_loss = 0.0

    for images, labels in balanced_train_loader:
        images = images.to(DEVICE)
        labels = labels.to(DEVICE)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        total_loss += loss.item() * images.size(0)

    # Evaluate on the same held-out validation set
    model.eval()
    true_labels = []
    predictions = []

    with torch.no_grad():
        for images, labels in val_loader:
            outputs = model(images.to(DEVICE))
            predictions.extend(outputs.argmax(dim=1).cpu().tolist())
            true_labels.extend(labels.tolist())

    macro_f1 = f1_score(
        true_labels,
        predictions,
        labels=list(range(len(classes))),
        average="macro",
        zero_division=0,
    )

    avg_loss = total_loss / len(train_dataset)
    print(
        f"Epoch {epoch + 1:02d}/{EPOCHS} | "
        f"Loss: {avg_loss:.4f} | "
        f"Validation macro F1: {macro_f1:.3f}"
    )

    if macro_f1 > best_macro_f1:
        best_macro_f1 = macro_f1
        best_state = {
            key: value.detach().cpu().clone()
            for key, value in model.state_dict().items()
        }

# Restore the best checkpoint from this experiment
if best_state is not None:
    model.load_state_dict(best_state)

model.eval()
final_true = []
final_pred = []

with torch.no_grad():
    for images, labels in val_loader:
        outputs = model(images.to(DEVICE))
        final_pred.extend(outputs.argmax(dim=1).cpu().tolist())
        final_true.extend(labels.tolist())

print("\n=== Balanced Training Results ===")
print("Best validation macro F1:", round(best_macro_f1, 4))
print(classification_report(
    final_true,
    final_pred,
    labels=list(range(len(classes))),
    target_names=classes,
    zero_division=0,
))

MODEL_PATH = "/kaggle/working/geovision_pilot_cnn_balanced.pth"

torch.save({
    "model_state_dict": model.state_dict(),
    "classes": classes,
    "input_size": 128,
    "best_validation_macro_f1": best_macro_f1,
}, MODEL_PATH)

print("Saved model:", MODEL_PATH)
```

```python

import json
import pandas as pd
from pathlib import Path
from collections import Counter
from PIL import Image
from shapely import wkt

BASE_DIR = Path("/kaggle/working/geovision_large_pilot")
OUTPUT_DIR = BASE_DIR / "crops"

manifests = {
    "train": BASE_DIR / "train_manifest.csv",
    "val": BASE_DIR / "val_manifest.csv",
}

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

all_counts = Counter()
all_rows = []
failed = 0
padding = 8

for split, manifest_path in manifests.items():
    df = pd.read_csv(manifest_path)
    split_rows = []

    print(f"Processing {split}: {len(df)} buildings")

    for _, row in df.iterrows():
        try:
            image_path = Path(row["image_path"])
            label_path = Path(row["annotation_path"])

            with label_path.open("r", encoding="utf-8") as f:
                annotation = json.load(f)

            target = None
            for building in annotation.get(
                "features", {}
            ).get("xy", []):
                properties = building.get("properties", {})
                if properties.get("uid") == row["building_id"]:
                    target = building
                    break

            if target is None or not target.get("wkt"):
                failed += 1
                continue

            polygon = wkt.loads(target["wkt"])
            min_x, min_y, max_x, max_y = polygon.bounds

            with Image.open(image_path) as source:
                image = source.convert("RGB")

            left = max(0, int(min_x) - padding)
            top = max(0, int(min_y) - padding)
            right = min(image.width, int(max_x) + padding + 1)
            bottom = min(image.height, int(max_y) + padding + 1)

            if right - left < 16 or bottom - top < 16:
                failed += 1
                continue

            crop = image.crop((left, top, right, bottom))

            label = row["damage_label"]
            class_dir = OUTPUT_DIR / split / label
            class_dir.mkdir(parents=True, exist_ok=True)

            filename = f"{row['building_id']}.png"
            crop_path = class_dir / filename
            crop.save(crop_path)

            record = {
                "crop_path": str(crop_path),
                "damage_label": label,
                "image_name": row["image_name"],
                "building_id": row["building_id"],
                "original_width": crop.width,
                "original_height": crop.height,
            }

            split_rows.append(record)
            all_rows.append({"split": split, **record})
            all_counts[f"{split}/{label}"] += 1

        except Exception as error:
            failed += 1
            print("Skipped:", row["building_id"], "|", str(error))

    pd.DataFrame(split_rows).to_csv(
        BASE_DIR / f"{split}_crops_manifest.csv",
        index=False,
    )

    print(f"Saved {len(split_rows)} {split} crops")

print("\n=== GeoVision Crop Extraction Report ===")
print("Total crops created:", len(all_rows))
print("Failed or skipped records:", failed)

print("\nCrops per split and class:")
for key, count in sorted(all_counts.items()):
    print(f"{key}: {count}")

print("\nOutput folder:", OUTPUT_DIR)
print("Training manifest:", BASE_DIR / "train_crops_manifest.csv")
print("Validation manifest:", BASE_DIR / "val_crops_manifest.csv")
```

```python

import pandas as pd
from pathlib import Path
from PIL import Image

BASE_DIR = Path("/kaggle/working/geovision_large_pilot")
OUTPUT_DIR = BASE_DIR / "crops_128"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

manifests = {
    "train": BASE_DIR / "train_crops_manifest.csv",
    "val": BASE_DIR / "val_crops_manifest.csv",
}

for split, manifest_path in manifests.items():
    df = pd.read_csv(manifest_path)
    resized_rows = []
    failures = 0

    print(f"Resizing {split}: {len(df)} crops")

    for _, row in df.iterrows():
        try:
            with Image.open(row["crop_path"]) as image:
                image = image.convert("RGB")
                original_width, original_height = image.size

                resized = image.resize(
                    (128, 128),
                    Image.Resampling.BILINEAR,
                )

                output_path = (
                    OUTPUT_DIR
                    / split
                    / row["damage_label"]
                    / Path(row["crop_path"]).name
                )
                output_path.parent.mkdir(
                    parents=True,
                    exist_ok=True,
                )
                resized.save(output_path)

            record = row.to_dict()
            record["resized_path"] = str(output_path)
            record["original_width"] = original_width
            record["original_height"] = original_height
            resized_rows.append(record)

        except Exception as error:
            failures += 1
            print("Failed:", row["crop_path"], "|", error)

    output_csv = BASE_DIR / f"{split}_crops_128_manifest.csv"
    pd.DataFrame(resized_rows).to_csv(output_csv, index=False)

    print(f"Successfully resized: {len(resized_rows)}")
    print(f"Failures: {failures}")
    print(f"Manifest: {output_csv}")

print("\nResize operation finished.")
```

```python
import os
import torch
import torch.nn as nn
import pandas as pd
import numpy as np

from PIL import Image
from torchvision import transforms
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score
)

BASE_DIR = "/kaggle/working/geovision_large_pilot"
MANIFEST = os.path.join(BASE_DIR, "val_crops_128_manifest.csv")
MODEL_PATH = "/kaggle/working/geovision_pilot_cnn_balanced.pth"

device = torch.device("cpu")
checkpoint = torch.load(
    MODEL_PATH, map_location=device, weights_only=True
)

classes = checkpoint["classes"]
print("Classes:", classes)

# Re-create the CNN architecture
class DamageCNN(nn.Module):
    def __init__(self, num_classes=4):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 16, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Conv2d(16, 32, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Conv2d(32, 64, 3, padding=1),
            nn.ReLU(),
            nn.AdaptiveAvgPool2d((1, 1))
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(32, num_classes)
        )

    def forward(self, x):
        return self.classifier(self.features(x))

model = DamageCNN(num_classes=len(classes)).to(device)
model.load_state_dict(checkpoint["model_state_dict"])
model.eval()

# Load validation images
df = pd.read_csv(MANIFEST)

transform = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.ToTensor()
])

class ValidationDataset(Dataset):
    def __init__(self, dataframe):
        self.df = dataframe.reset_index(drop=True)

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        image = Image.open(row["resized_path"]).convert("RGB")
        image = transform(image)
        label = classes.index(row["damage_label"])
        return image, label

dataset = ValidationDataset(df)
loader = DataLoader(dataset, batch_size=32, shuffle=False)

y_true, y_pred = [], []

with torch.inference_mode():
    for images, labels in loader:
        outputs = model(images.to(device))
        predictions = outputs.argmax(dim=1)
        y_true.extend(labels.numpy())
        y_pred.extend(predictions.cpu().numpy())

print("\nValidation accuracy:", round(accuracy_score(y_true, y_pred), 4))
print("Macro F1:", round(f1_score(
    y_true, y_pred, average="macro", zero_division=0
), 4))

print("\nClassification report:")
print(classification_report(
    y_true, y_pred, labels=list(range(len(classes))),
    target_names=classes, zero_division=0
))

print("Confusion matrix (rows = actual, columns = predicted):")
print(confusion_matrix(y_true, y_pred, labels=list(range(len(classes)))))
```

```python
import numpy as np
import torch
from torch import nn
from torch.utils.data import DataLoader, WeightedRandomSampler
from sklearn.metrics import f1_score, classification_report

# Reproducibility
torch.manual_seed(42)
np.random.seed(42)

DEVICE = torch.device("cpu")

# Ensure datasets use the current dataframes
train_dataset = BuildingCropDataset(train_df)
val_dataset = BuildingCropDataset(val_df)

# Check label mapping
class_to_id = {name: i for i, name in enumerate(classes)}
train_df["label_id"] = train_df["damage_label"].map(class_to_id)
val_df["label_id"] = val_df["damage_label"].map(class_to_id)

# Balanced sampling: give each class similar sampling probability
train_labels = train_df["label_id"].to_numpy()
class_counts = np.bincount(train_labels, minlength=len(classes))
sample_weights = 1.0 / class_counts[train_labels]

sampler = WeightedRandomSampler(
    weights=torch.as_tensor(sample_weights, dtype=torch.double),
    num_samples=len(train_labels),
    replacement=True,
)

balanced_train_loader = DataLoader(
    train_dataset,
    batch_size=16,
    sampler=sampler,
    num_workers=0,
)

val_loader = DataLoader(
    val_dataset,
    batch_size=16,
    shuffle=False,
    num_workers=0,
)

# CNN architecture
class DamageCNN(nn.Module):
    def __init__(self, num_classes=4):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 16, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Conv2d(16, 32, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Conv2d(32, 64, 3, padding=1),
            nn.ReLU(),
            nn.AdaptiveAvgPool2d((1, 1)),
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(32, num_classes),
        )

    def forward(self, x):
        x = self.features(x)
        return self.classifier(x)


model = DamageCNN(num_classes=len(classes)).to(DEVICE)

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.0005,
    weight_decay=0.0001,
)

EPOCHS = 15
best_macro_f1 = -1.0
best_state = None

print("Training samples:", len(train_dataset))
print("Validation samples:", len(val_dataset))
print("Classes:", classes)
print("\nStarting training...\n")

for epoch in range(EPOCHS):
    model.train()
    total_loss = 0.0

    for images, labels in balanced_train_loader:
        images = images.to(DEVICE)
        labels = labels.to(DEVICE)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        total_loss += loss.item() * images.size(0)

    # Evaluate on validation data
    model.eval()
    y_true = []
    y_pred = []

    with torch.no_grad():
        for images, labels in val_loader:
            outputs = model(images.to(DEVICE))
            predictions = outputs.argmax(dim=1)

            y_true.extend(labels.tolist())
            y_pred.extend(predictions.cpu().tolist())

    macro_f1 = f1_score(
        y_true,
        y_pred,
        labels=list(range(len(classes))),
        average="macro",
        zero_division=0,
    )

    avg_loss = total_loss / len(train_dataset)

    print(
        f"Epoch {epoch + 1:02d}/{EPOCHS} | "
        f"Loss: {avg_loss:.4f} | "
        f"Validation macro F1: {macro_f1:.4f}"
    )

    if macro_f1 > best_macro_f1:
        best_macro_f1 = macro_f1
        best_state = {
            k: v.detach().cpu().clone()
            for k, v in model.state_dict().items()
        }

# Restore the best model
if best_state is not None:
    model.load_state_dict(best_state)

model.eval()

# Final evaluation
final_true = []
final_pred = []

with torch.no_grad():
    for images, labels in val_loader:
        outputs = model(images.to(DEVICE))
        final_pred.extend(outputs.argmax(dim=1).cpu().tolist())
        final_true.extend(labels.tolist())

print("\n=== Final Balanced CNN Results ===")
print("Best validation macro F1:", round(best_macro_f1, 4))
print(
    classification_report(
        final_true,
        final_pred,
        labels=list(range(len(classes))),
        target_names=classes,
        zero_division=0,
    )
)

# Save the best checkpoint
MODEL_PATH = "/kaggle/working/geovision_pilot_cnn_balanced.pth"

torch.save(
    {
        "model_state_dict": model.state_dict(),
        "classes": classes,
        "input_size": 128,
        "best_validation_macro_f1": best_macro_f1,
    },
    MODEL_PATH,
)

print("\nSaved model:", MODEL_PATH)
```

```python

import os
import json
import random
import pandas as pd
from shapely import wkt

DATASET_PATH = (
    "/kaggle/input/datasets/tunguz/"
    "xview2-challenge-dataset-train-and-test"
)
IMAGE_DIR = os.path.join(DATASET_PATH, "train", "train", "images")
LABEL_DIR = os.path.join(DATASET_PATH, "train", "train", "labels")

OUTPUT_DIR = "/kaggle/working/geovision_large_pilot"
os.makedirs(OUTPUT_DIR, exist_ok=True)

CLASSES = ["no-damage", "minor-damage", "major-damage", "destroyed"]
SAMPLES_PER_CLASS = 2000
SEED = 42
random.seed(SEED)

records = []

annotation_files = sorted(
    f for f in os.listdir(LABEL_DIR)
    if f.endswith("_post_disaster.json")
)

for i, filename in enumerate(annotation_files):
    label_path = os.path.join(LABEL_DIR, filename)
    image_name = filename.replace(".json", ".png")
    image_path = os.path.join(IMAGE_DIR, image_name)

    if not os.path.exists(image_path):
        continue

    with open(label_path, "r", encoding="utf-8") as file:
        data = json.load(file)

    for feature in data.get("features", {}).get("xy", []):
        props = feature.get("properties", {})
        label = props.get("subtype")

        if props.get("feature_type") != "building" or label not in CLASSES:
            continue

        try:
            polygon = wkt.loads(feature["wkt"])
            min_x, min_y, max_x, max_y = polygon.bounds

            records.append({
                "image_name": image_name,
                "image_path": image_path,
                "building_id": props.get("uid", ""),
                "damage_label": label,
                "min_x": min_x,
                "min_y": min_y,
                "max_x": max_x,
                "max_y": max_y,
            })
        except Exception:
            continue

    if (i + 1) % 500 == 0:
        print(f"Processed {i + 1}/{len(annotation_files)} annotation files")

df = pd.DataFrame(records)
print("\nAll eligible buildings:", len(df))
print("Available classes:\n", df["damage_label"].value_counts())

# Sample up to 2,000 buildings per class.
sampled_parts = []
for label in CLASSES:
    class_df = df[df["damage_label"] == label]
    n = min(SAMPLES_PER_CLASS, len(class_df))
    sampled_parts.append(class_df.sample(n=n, random_state=SEED))

sampled = pd.concat(sampled_parts, ignore_index=True)

# Split by image, never by individual building.
image_names = sampled["image_name"].drop_duplicates().tolist()
random.Random(SEED).shuffle(image_names)

split_index = int(len(image_names) * 0.80)
train_images = set(image_names[:split_index])
val_images = set(image_names[split_index:])

train_df = sampled[sampled["image_name"].isin(train_images)].copy()
val_df = sampled[sampled["image_name"].isin(val_images)].copy()

train_manifest = os.path.join(OUTPUT_DIR, "train_manifest.csv")
val_manifest = os.path.join(OUTPUT_DIR, "val_manifest.csv")

train_df.to_csv(train_manifest, index=False)
val_df.to_csv(val_manifest, index=False)

print("\nTrain buildings:", len(train_df), "| images:", train_df["image_name"].nunique())
print("Validation buildings:", len(val_df), "| images:", val_df["image_name"].nunique())
print("Image overlap:", len(train_images & val_images))
print("\nTrain class counts:\n", train_df["damage_label"].value_counts())
print("\nValidation class counts:\n", val_df["damage_label"].value_counts())
print("\nSaved manifests to:", OUTPUT_DIR)
```

```python

import os
import pandas as pd
from PIL import Image
from collections import Counter

OUTPUT_DIR = "/kaggle/working/geovision_large_pilot"
CROP_SIZE = 128
MARGIN = 8

CLASS_NAMES = ["no-damage", "minor-damage", "major-damage", "destroyed"]

def create_crops(manifest_path, split):
    df = pd.read_csv(manifest_path)
    crop_records = []
    failed = 0

    for label in CLASS_NAMES:
        os.makedirs(
            os.path.join(OUTPUT_DIR, "crops_128", split, label),
            exist_ok=True
        )

    for i, row in df.iterrows():
        try:
            with Image.open(row["image_path"]) as original:
                image = original.convert("RGB")
                width, height = image.size

                x1 = max(0, int(row["min_x"]) - MARGIN)
                y1 = max(0, int(row["min_y"]) - MARGIN)
                x2 = min(width, int(row["max_x"]) + MARGIN + 1)
                y2 = min(height, int(row["max_y"]) + MARGIN + 1)

                if x2 <= x1 or y2 <= y1:
                    failed += 1
                    continue

                crop = image.crop((x1, y1, x2, y2))
                crop = crop.resize((CROP_SIZE, CROP_SIZE))

                building_id = str(row["building_id"])
                if not building_id or building_id == "nan":
                    building_id = f"row_{i}"

                crop_name = (
                    f"{os.path.splitext(row['image_name'])[0]}_"
                    f"{building_id}.png"
                )
                crop_path = os.path.join(
                    OUTPUT_DIR, "crops_128", split,
                    row["damage_label"], crop_name
                )
                crop.save(crop_path)

                crop_records.append({
                    "resized_path": crop_path,
                    "damage_label": row["damage_label"],
                    "image_name": row["image_name"],
                    "building_id": building_id,
                    "original_width": width,
                    "original_height": height,
                })

        except Exception:
            failed += 1

        if (i + 1) % 1000 == 0:
            print(f"{split}: processed {i + 1}/{len(df)} buildings")

    result = pd.DataFrame(crop_records)
    result_path = os.path.join(
        OUTPUT_DIR, f"{split}_crops_128_manifest.csv"
    )
    result.to_csv(result_path, index=False)

    print(f"\n{split} crops created:", len(result))
    print(f"{split} failed:", failed)
    print("Class counts:\n", result["damage_label"].value_counts())
    print("Manifest:", result_path)

    return result

train_crops_df = create_crops(
    os.path.join(OUTPUT_DIR, "train_manifest.csv"), "train"
)

val_crops_df = create_crops(
    os.path.join(OUTPUT_DIR, "val_manifest.csv"), "val"
)
```

```python

import os
import random
import numpy as np
import pandas as pd
import torch
from PIL import Image
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from torchvision import transforms

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

CLASSES = ["no-damage", "minor-damage", "major-damage", "destroyed"]
CLASS_TO_ID = {name: i for i, name in enumerate(CLASSES)}

OUTPUT_DIR = "/kaggle/working/geovision_large_pilot"

train_df = pd.read_csv(
    os.path.join(OUTPUT_DIR, "train_crops_128_manifest.csv")
)
val_df = pd.read_csv(
    os.path.join(OUTPUT_DIR, "val_crops_128_manifest.csv")
)

train_df["label_id"] = train_df["damage_label"].map(CLASS_TO_ID)
val_df["label_id"] = val_df["damage_label"].map(CLASS_TO_ID)

assert train_df["label_id"].notna().all()
assert val_df["label_id"].notna().all()

transform = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.ToTensor(),
])

class BuildingCropDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.df = dataframe.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, index):
        row = self.df.iloc[index]
        image = Image.open(row["resized_path"]).convert("RGB")

        if self.transform:
            image = self.transform(image)

        label = int(row["label_id"])
        return image, label

train_dataset = BuildingCropDataset(train_df, transform)
val_dataset = BuildingCropDataset(val_df, transform)

# Give less frequent classes higher sampling probability.
class_counts = train_df["label_id"].value_counts().sort_index()
class_weights = 1.0 / class_counts.to_numpy()
sample_weights = train_df["label_id"].map(
    dict(enumerate(class_weights))
).to_numpy()

sampler = WeightedRandomSampler(
    weights=torch.as_tensor(sample_weights, dtype=torch.double),
    num_samples=len(sample_weights),
    replacement=True
)

BATCH_SIZE = 32

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    sampler=sampler,
    num_workers=0
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)

print("Classes:", CLASSES)
print("Training samples:", len(train_dataset))
print("Validation samples:", len(val_dataset))
print("Training batches:", len(train_loader))
print("Validation batches:", len(val_loader))

images, labels = next(iter(train_loader))
print("Sample image tensor shape:", images.shape)
print("Sample label tensor shape:", labels.shape)
print("Sample labels:", labels[:10].tolist())
```

```python

import os
import pandas as pd
from PIL import Image

OUTPUT_DIR = "/kaggle/working/geovision_large_pilot"
CROP_SIZE = 128
MARGIN = 8

CLASSES = ["no-damage", "minor-damage", "major-damage", "destroyed"]

def create_crops(manifest_path, split):
    df = pd.read_csv(manifest_path)
    crop_records = []
    failed = 0

    for label in CLASSES:
        os.makedirs(
            os.path.join(OUTPUT_DIR, "crops_128", split, label),
            exist_ok=True
        )

    # Reuse each source image while processing its buildings.
    for image_name, group in df.groupby("image_name", sort=False):
        try:
            image_path = group["image_path"].iloc[0]

            with Image.open(image_path) as original:
                image = original.convert("RGB")
                width, height = image.size

                for i, row in group.iterrows():
                    try:
                        x1 = max(0, int(row["min_x"]) - MARGIN)
                        y1 = max(0, int(row["min_y"]) - MARGIN)
                        x2 = min(width, int(row["max_x"]) + MARGIN + 1)
                        y2 = min(height, int(row["max_y"]) + MARGIN + 1)

                        if x2 <= x1 or y2 <= y1:
                            failed += 1
                            continue

                        crop = image.crop((x1, y1, x2, y2))
                        crop = crop.resize((CROP_SIZE, CROP_SIZE))

                        building_id = str(row["building_id"])
                        if not building_id or building_id == "nan":
                            building_id = f"row_{i}"

                        crop_name = (
                            f"{os.path.splitext(image_name)[0]}_"
                            f"{building_id}.png"
                        )

                        crop_path = os.path.join(
                            OUTPUT_DIR, "crops_128", split,
                            row["damage_label"], crop_name
                        )

                        crop.save(crop_path)

                        crop_records.append({
                            "resized_path": crop_path,
                            "damage_label": row["damage_label"],
                            "image_name": image_name,
                            "building_id": building_id,
                            "original_width": width,
                            "original_height": height,
                        })

                    except Exception as error:
                        failed += 1

        except Exception as error:
            failed += len(group)
            print("Could not process image:", image_name, "|", str(error))

        if len(crop_records) and len(crop_records) % 1000 < len(group):
            print(
                f"{split}: created {len(crop_records)} crops "
                f"out of {len(df)} buildings"
            )

    result = pd.DataFrame(crop_records)

    result_path = os.path.join(
        OUTPUT_DIR, f"{split}_crops_128_manifest.csv"
    )
    result.to_csv(result_path, index=False)

    print(f"\n{split} crops created:", len(result))
    print(f"{split} failed:", failed)

    if not result.empty:
        print("Class counts:")
        print(result["damage_label"].value_counts())

    print("Manifest saved:", result_path)

    return result

train_crops_df = create_crops(
    os.path.join(OUTPUT_DIR, "train_manifest.csv"),
    "train"
)

val_crops_df = create_crops(
    os.path.join(OUTPUT_DIR, "val_manifest.csv"),
    "val"
)

print("\nAll crop generation finished.")

```

```python

import os
import random
import numpy as np
import pandas as pd
import torch
from PIL import Image
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from torchvision import transforms

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

CLASSES = ["no-damage", "minor-damage", "major-damage", "destroyed"]
CLASS_TO_ID = {name: i for i, name in enumerate(CLASSES)}

OUTPUT_DIR = "/kaggle/working/geovision_large_pilot"

train_df = pd.read_csv(
    os.path.join(OUTPUT_DIR, "train_crops_128_manifest.csv")
)
val_df = pd.read_csv(
    os.path.join(OUTPUT_DIR, "val_crops_128_manifest.csv")
)

train_df["label_id"] = train_df["damage_label"].map(CLASS_TO_ID)
val_df["label_id"] = val_df["damage_label"].map(CLASS_TO_ID)

assert train_df["label_id"].notna().all()
assert val_df["label_id"].notna().all()

transform = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.ToTensor(),
])

class BuildingCropDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.df = dataframe.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, index):
        row = self.df.iloc[index]

        with Image.open(row["resized_path"]) as image_file:
            image = image_file.convert("RGB")

        if self.transform:
            image = self.transform(image)

        return image, int(row["label_id"])

train_dataset = BuildingCropDataset(train_df, transform)
val_dataset = BuildingCropDataset(val_df, transform)

# Weighted sampling reduces the effect of class imbalance.
class_counts = train_df["label_id"].value_counts().sort_index()
class_weights = 1.0 / class_counts.to_numpy()

sample_weights = train_df["label_id"].map(
    dict(enumerate(class_weights))
).to_numpy()

sampler = WeightedRandomSampler(
    weights=torch.as_tensor(sample_weights, dtype=torch.double),
    num_samples=len(sample_weights),
    replacement=True,
)

BATCH_SIZE = 32

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    sampler=sampler,
    num_workers=0,
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
)

print("PyTorch:", torch.__version__)
print("Device:", "cuda" if torch.cuda.is_available() else "cpu")
print("Classes:", CLASSES)
print("Training samples:", len(train_dataset))
print("Validation samples:", len(val_dataset))
print("Training batches:", len(train_loader))
print("Validation batches:", len(val_loader))

images, labels = next(iter(train_loader))
print("Sample image shape:", images.shape)
print("Sample label shape:", labels.shape)
print("Sample labels:", labels[:10].tolist())
```

```python

import torch
import torch.nn as nn

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

class DamageCNN(nn.Module):
    def __init__(self, num_classes=4):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(3, 16, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),

            nn.AdaptiveAvgPool2d((1, 1)),
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(32, num_classes),
        )

    def forward(self, x):
        return self.classifier(self.features(x))

model = DamageCNN(num_classes=len(CLASSES)).to(DEVICE)

# Test one batch.
images, labels = next(iter(train_loader))
images = images.to(DEVICE)

with torch.no_grad():
    outputs = model(images)

print("Device:", DEVICE)
print("Model output shape:", outputs.shape)
print("Trainable parameters:", sum(
    p.numel() for p in model.parameters() if p.requires_grad
))
print("GPU:", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "Not available")
```

```python

import os
import copy
import time
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from sklearn.metrics import f1_score

OUTPUT_DIR = "/kaggle/working/geovision_large_pilot"
MODEL_PATH = "/kaggle/working/geovision_pilot_cnn_balanced.pth"
HISTORY_PATH = os.path.join(OUTPUT_DIR, "training_history.csv")

EPOCHS = 15
LEARNING_RATE = 0.001

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=LEARNING_RATE
)

best_val_f1 = -1.0
best_state = None
history = []

print("Training on:", DEVICE)
print("Training samples:", len(train_dataset))
print("Validation samples:", len(val_dataset))
print("-" * 65)

for epoch in range(EPOCHS):
    start_time = time.time()

    # Training
    model.train()
    running_loss = 0.0
    total_samples = 0

    for images, labels in train_loader:
        images = images.to(DEVICE)
        labels = labels.to(DEVICE)

        optimizer.zero_grad(set_to_none=True)

        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * labels.size(0)
        total_samples += labels.size(0)

    train_loss = running_loss / total_samples

    # Validation
    model.eval()
    all_predictions = []
    all_labels = []

    with torch.no_grad():
        for images, labels in val_loader:
            images = images.to(DEVICE)
            outputs = model(images)

            predictions = outputs.argmax(dim=1).cpu().numpy()
            all_predictions.extend(predictions)
            all_labels.extend(labels.numpy())

    val_macro_f1 = f1_score(
        all_labels,
        all_predictions,
        labels=list(range(len(CLASSES))),
        average="macro",
        zero_division=0
    )

    elapsed = time.time() - start_time

    history.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "val_macro_f1": val_macro_f1,
        "seconds": elapsed,
    })

    print(
        f"Epoch {epoch + 1:02d}/{EPOCHS} | "
        f"Loss: {train_loss:.4f} | "
        f"Validation macro F1: {val_macro_f1:.4f} | "
        f"Time: {elapsed:.1f}s"
    )

    # Keep the best-performing model.
    if val_macro_f1 > best_val_f1:
        best_val_f1 = val_macro_f1
        best_state = copy.deepcopy(model.state_dict())

        torch.save({
            "model_state_dict": best_state,
            "classes": CLASSES,
            "input_size": 128,
            "best_validation_macro_f1": best_val_f1,
            "epoch": epoch + 1,
            "architecture": "DamageCNN",
        }, MODEL_PATH)

        print("  -> Best checkpoint saved.")

# Restore best model weights and save the complete history.
model.load_state_dict(best_state)

history_df = pd.DataFrame(history)
history_df.to_csv(HISTORY_PATH, index=False)

print("\nTraining complete.")
print("Best validation macro F1:", round(best_val_f1, 4))
print("Best model:", MODEL_PATH)
print("History:", HISTORY_PATH)
print("Checkpoint exists:", os.path.exists(MODEL_PATH))
```

```python
import cv2
import numpy as np
from PIL import Image
from torch.utils.data import Dataset, DataLoader
import torch

class SharpenedValidationDataset(Dataset):
    def __init__(self, original_dataset, amount=0.5):
        self.original_dataset = original_dataset
        self.amount = amount

    def __len__(self):
        return len(self.original_dataset)

    def __getitem__(self, idx):
        image, label = self.original_dataset[idx]

        # Convert normalized tensor back to image pixels
        image_np = image.permute(1, 2, 0).cpu().numpy()

        mean = np.array([0.485, 0.456, 0.406])
        std = np.array([0.229, 0.224, 0.225])

        image_np = image_np * std + mean
        image_np = np.clip(image_np, 0, 1)

        image_uint8 = (image_np * 255).astype(np.uint8)

        # Mild sharpening using an unsharp mask
        blurred = cv2.GaussianBlur(
            image_uint8, (0, 0), sigmaX=1.0
        )

        sharpened = cv2.addWeighted(
            image_uint8, 1.0 + self.amount,
            blurred, -self.amount, 0
        )

        sharpened = np.clip(sharpened, 0, 255).astype(np.uint8)

        # Convert back to normalized tensor
        sharpened = torch.from_numpy(
            sharpened.astype(np.float32) / 255.0
        ).permute(2, 0, 1)

        sharpened = (
            sharpened - torch.tensor(mean, dtype=torch.float32)[:, None, None]
        ) / torch.tensor(std, dtype=torch.float32)[:, None, None]

        return sharpened, label


sharpened_val_dataset = SharpenedValidationDataset(
    val_dataset, amount=0.5
)

sharpened_val_loader = DataLoader(
    sharpened_val_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=2
)

print("Original validation images:", len(val_dataset))
print("Sharpened validation images:", len(sharpened_val_dataset))
print("Sharpening strength:", 0.5)
print("Original dataset and checkpoint remain unchanged.")
```

```python
import os
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import models

# 1. Select device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 2. Load pretrained ResNet18
weights = models.ResNet18_Weights.DEFAULT
resnet_new = models.resnet18(weights=weights)

# 3. Replace final layer for 4 damage classes
resnet_new.fc = nn.Linear(
    resnet_new.fc.in_features,
    4
)

# 4. Freeze all parameters
for param in resnet_new.parameters():
    param.requires_grad = False

# 5. Unfreeze layer3, layer4, and classifier
for param in resnet_new.layer3.parameters():
    param.requires_grad = True

for param in resnet_new.layer4.parameters():
    param.requires_grad = True

for param in resnet_new.fc.parameters():
    param.requires_grad = True

# 6. Move model to GPU/CPU
resnet_new = resnet_new.to(device)

# 7. Define loss function
criterion_new = nn.CrossEntropyLoss()

# 8. Define optimizer
optimizer_new = optim.AdamW(
    filter(
        lambda p: p.requires_grad,
        resnet_new.parameters()
    ),
    lr=1e-4,
    weight_decay=1e-4
)

# 9. New checkpoint path
# This does NOT overwrite your previous best checkpoint
checkpoint_path_new = (
    "/kaggle/working/geovision_resnet18_unseen_disaster.pth"
)

# 10. Display model setup information
trainable_params = sum(
    p.numel()
    for p in resnet_new.parameters()
    if p.requires_grad
)

print("MODEL READY")
print("=" * 40)
print("Device:", device)
print("Trainable parameters:", trainable_params)
print("Training batches:", len(train_loader_new))
print("Validation batches:", len(val_loader_new))
print("Test batches:", len(test_loader_new))
print("New checkpoint:", checkpoint_path_new)
```

```python

import copy
import torch
import numpy as np
from sklearn.metrics import accuracy_score, f1_score

EPOCHS = 10
PATIENCE = 3

best_val_f1 = -1.0
best_epoch = 0
epochs_without_improvement = 0

history_new = []

print("GeoVision — Unseen Disaster Training")
print("=" * 55)

for epoch in range(EPOCHS):
    # -------- TRAINING --------
    resnet_new.train()

    train_losses = []
    train_true = []
    train_pred = []

    for images, labels in train_loader_new:
        images = images.to(device)
        labels = labels.to(device)

        optimizer_new.zero_grad()

        outputs = resnet_new(images)
        loss = criterion_new(outputs, labels)

        loss.backward()
        optimizer_new.step()

        train_losses.append(loss.item())

        predictions = outputs.argmax(dim=1)
        train_true.extend(labels.cpu().numpy())
        train_pred.extend(predictions.cpu().numpy())

    train_loss = float(np.mean(train_losses))
    train_acc = accuracy_score(train_true, train_pred)

    # -------- VALIDATION --------
    resnet_new.eval()

    val_losses = []
    val_true = []
    val_pred = []

    with torch.no_grad():
        for images, labels in val_loader_new:
            images = images.to(device)
            labels = labels.to(device)

            outputs = resnet_new(images)
            loss = criterion_new(outputs, labels)

            val_losses.append(loss.item())

            predictions = outputs.argmax(dim=1)
            val_true.extend(labels.cpu().numpy())
            val_pred.extend(predictions.cpu().numpy())

    val_loss = float(np.mean(val_losses))
    val_acc = accuracy_score(val_true, val_pred)
    val_f1 = f1_score(
        val_true,
        val_pred,
        average="macro",
        zero_division=0
    )

    history_new.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "train_accuracy": train_acc,
        "val_loss": val_loss,
        "val_accuracy": val_acc,
        "val_macro_f1": val_f1
    })

    print(
        f"Epoch {epoch + 1:02d}/{EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc:.4f} | "
        f"Val Macro F1: {val_f1:.4f}"
    )

    # -------- SAVE BEST MODEL --------
    if val_f1 > best_val_f1:
        best_val_f1 = val_f1
        best_epoch = epoch + 1
        epochs_without_improvement = 0

        best_state_new = copy.deepcopy(resnet_new.state_dict())

        torch.save({
            "epoch": best_epoch,
            "model_state_dict": best_state_new,
            "optimizer_state_dict": optimizer_new.state_dict(),
            "val_macro_f1": best_val_f1,
            "class_to_idx": class_to_idx
        }, checkpoint_path_new)

        print("  -> Best checkpoint saved.")
    else:
        epochs_without_improvement += 1

    # Early stopping
    if epochs_without_improvement >= PATIENCE:
        print(f"\nEarly stopping at epoch {epoch + 1}.")
        break

print("\nTRAINING COMPLETE")
print("=" * 55)
print("Best epoch:", best_epoch)
print(f"Best validation Macro F1: {best_val_f1:.4f}")
print("Checkpoint:", checkpoint_path_new)
```

```python

import torch
import numpy as np
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# 1. Load the best checkpoint
checkpoint = torch.load(
    checkpoint_path_new,
    map_location=device,
    weights_only=False
)

resnet_new.load_state_dict(checkpoint["model_state_dict"])
resnet_new.eval()

print("Loaded best checkpoint")
print("Best epoch:", checkpoint["epoch"])
print("Best validation Macro F1:", round(checkpoint["val_macro_f1"], 4))

# 2. Evaluate on the held-out test set
test_true = []
test_pred = []

with torch.no_grad():
    for images, labels in test_loader_new:
        images = images.to(device)

        outputs = resnet_new(images)
        predictions = outputs.argmax(dim=1)

        test_true.extend(labels.numpy())
        test_pred.extend(predictions.cpu().numpy())

# 3. Calculate metrics
test_acc = accuracy_score(test_true, test_pred)
test_macro_f1 = f1_score(
    test_true,
    test_pred,
    average="macro",
    zero_division=0
)

print("\nFINAL HELD-OUT TEST RESULTS")
print("=" * 55)
print(f"Test samples: {len(test_true)}")
print(f"Test Accuracy: {test_acc:.4f}")
print(f"Test Macro F1: {test_macro_f1:.4f}")

# 4. Per-class performance
print("\nCLASSIFICATION REPORT")
print(classification_report(
    test_true,
    test_pred,
    labels=[0, 1, 2, 3],
    target_names=class_names,
    digits=4,
    zero_division=0
))

# 5. Confusion matrix
print("CONFUSION MATRIX")
print("Rows = actual class; Columns = predicted class")
print(confusion_matrix(
    test_true,
    test_pred,
    labels=[0, 1, 2, 3]
))
```

```python

import copy
import numpy as np
import torch
from sklearn.metrics import accuracy_score, f1_score

EPOCHS_WEIGHTED = 10
PATIENCE_WEIGHTED = 3

best_weighted_f1 = -1.0
best_weighted_epoch = 0
no_improvement_weighted = 0
history_weighted = []

print("GeoVision — Class-Weighted Training")
print("=" * 55)

for epoch in range(EPOCHS_WEIGHTED):
    # -------- TRAIN --------
    resnet_weighted.train()
    train_losses = []
    train_true = []
    train_pred = []

    for images, labels in train_loader_new:
        images = images.to(device)
        labels = labels.to(device)

        optimizer_weighted.zero_grad()
        outputs = resnet_weighted(images)
        loss = criterion_weighted(outputs, labels)

        loss.backward()
        optimizer_weighted.step()

        train_losses.append(loss.item())
        train_pred.extend(outputs.argmax(dim=1).detach().cpu().numpy())
        train_true.extend(labels.detach().cpu().numpy())

    train_loss = float(np.mean(train_losses))
    train_acc = accuracy_score(train_true, train_pred)

    # -------- VALIDATE --------
    resnet_weighted.eval()
    val_losses = []
    val_true = []
    val_pred = []

    with torch.no_grad():
        for images, labels in val_loader_new:
            images = images.to(device)
            labels = labels.to(device)

            outputs = resnet_weighted(images)
            loss = criterion_weighted(outputs, labels)

            val_losses.append(loss.item())
            val_pred.extend(outputs.argmax(dim=1).cpu().numpy())
            val_true.extend(labels.cpu().numpy())

    val_loss = float(np.mean(val_losses))
    val_acc = accuracy_score(val_true, val_pred)
    val_f1 = f1_score(
        val_true,
        val_pred,
        average="macro",
        zero_division=0
    )

    history_weighted.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "train_accuracy": train_acc,
        "val_loss": val_loss,
        "val_accuracy": val_acc,
        "val_macro_f1": val_f1
    })

    print(
        f"Epoch {epoch+1:02d}/{EPOCHS_WEIGHTED} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc:.4f} | "
        f"Val Macro F1: {val_f1:.4f}"
    )

    if val_f1 > best_weighted_f1:
        best_weighted_f1 = val_f1
        best_weighted_epoch = epoch + 1
        no_improvement_weighted = 0

        best_state_weighted = copy.deepcopy(
            resnet_weighted.state_dict()
        )

        torch.save({
            "epoch": best_weighted_epoch,
            "model_state_dict": best_state_weighted,
            "optimizer_state_dict": optimizer_weighted.state_dict(),
            "val_macro_f1": best_weighted_f1,
            "class_to_idx": class_to_idx
        }, checkpoint_weighted)

        print("  -> Best weighted checkpoint saved.")
    else:
        no_improvement_weighted += 1

    if no_improvement_weighted >= PATIENCE_WEIGHTED:
        print(f"\nEarly stopping at epoch {epoch+1}.")
        break

print("\nWEIGHTED TRAINING COMPLETE")
print("=" * 55)
print("Best epoch:", best_weighted_epoch)
print(f"Best validation Macro F1: {best_weighted_f1:.4f}")
print("Checkpoint:", checkpoint_weighted)
```

```python
from sklearn.metrics import f1_score
import copy

best_val_f1_aug = -1.0
best_epoch_aug = 0
patience_aug = 3
epochs_no_improve_aug = 0
num_epochs_aug = 10

for epoch in range(num_epochs_aug):
    # Training
    resnet_aug.train()
    running_loss = 0.0
    train_correct = 0
    train_total = 0

    for images, labels in train_loader_aug:
        images = images.to(device)
        labels = labels.to(device)

        optimizer_aug.zero_grad()

        outputs = resnet_aug(images)
        loss = criterion_aug(outputs, labels)

        loss.backward()
        optimizer_aug.step()

        running_loss += loss.item() * images.size(0)
        predictions = outputs.argmax(dim=1)
        train_correct += (predictions == labels).sum().item()
        train_total += labels.size(0)

    train_loss = running_loss / train_total
    train_acc = train_correct / train_total

    # Validation — no augmentation
    resnet_aug.eval()
    val_predictions = []
    val_labels_all = []
    val_loss_sum = 0.0
    val_total = 0

    with torch.no_grad():
        for images, labels in val_loader_new:
            images = images.to(device)
            labels = labels.to(device)

            outputs = resnet_aug(images)
            loss = criterion_aug(outputs, labels)

            val_loss_sum += loss.item() * images.size(0)
            predictions = outputs.argmax(dim=1)

            val_predictions.extend(predictions.cpu().numpy())
            val_labels_all.extend(labels.cpu().numpy())
            val_total += labels.size(0)

    val_loss = val_loss_sum / val_total
    val_acc = sum(
        p == y for p, y in zip(val_predictions, val_labels_all)
    ) / val_total

    val_macro_f1 = f1_score(
        val_labels_all,
        val_predictions,
        average="macro",
        zero_division=0
    )

    print(
        f"Epoch {epoch + 1}/{num_epochs_aug} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_acc:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_acc:.4f} | "
        f"Val Macro F1: {val_macro_f1:.4f}"
    )

    # Save the best checkpoint based on validation Macro F1
    if val_macro_f1 > best_val_f1_aug:
        best_val_f1_aug = val_macro_f1
        best_epoch_aug = epoch + 1
        epochs_no_improve_aug = 0

        torch.save(resnet_aug.state_dict(), checkpoint_aug)
        print("  Best checkpoint saved.")
    else:
        epochs_no_improve_aug += 1

    if epochs_no_improve_aug >= patience_aug:
        print("Early stopping triggered.")
        break

print("\nAugmented training complete.")
print("Best epoch:", best_epoch_aug)
print(f"Best validation Macro F1: {best_val_f1_aug:.4f}")
print("Checkpoint:", checkpoint_aug)
```

```python
import os
import pandas as pd
from PIL import Image

SOURCE_ROOT = (
    "/kaggle/input/datasets/tunguz/"
    "xview2-challenge-dataset-train-and-test/"
    "train/train/images"
)

OUTPUT_ROOT = "/kaggle/working/geovision_large_pilot"
CROP_SIZE = 128

os.makedirs(OUTPUT_ROOT, exist_ok=True)

def create_crops(manifest, split_name):
    output_rows = []
    crop_dir = os.path.join(OUTPUT_ROOT, "crops_128", split_name)
    os.makedirs(crop_dir, exist_ok=True)

    for i, row in manifest.iterrows():
        image_path = os.path.join(SOURCE_ROOT, row["image_name"])

        with Image.open(image_path) as img:
            img = img.convert("RGB")
            width, height = img.size

            # Clamp bounding box to image boundaries
            x1 = max(0, min(width, int(row["min_x"])))
            y1 = max(0, min(height, int(row["min_y"])))
            x2 = max(0, min(width, int(row["max_x"] + 0.999)))
            y2 = max(0, min(height, int(row["max_y"] + 0.999)))

            if x2 <= x1 or y2 <= y1:
                continue

            crop = img.crop((x1, y1, x2, y2))
            crop = crop.resize((CROP_SIZE, CROP_SIZE))

            label = str(row["damage_label"])
            label_dir = os.path.join(crop_dir, label)
            os.makedirs(label_dir, exist_ok=True)

            filename = (
                f"{os.path.splitext(row['image_name'])[0]}_"
                f"{row['building_id']}.png"
            )
            save_path = os.path.join(label_dir, filename)
            crop.save(save_path)

            output_rows.append({
                "resized_path": save_path,
                "damage_label": label,
                "image_name": row["image_name"],
                "building_id": row["building_id"],
                "original_width": width,
                "original_height": height
            })

        if (i + 1) % 500 == 0:
            print(f"{split_name}: processed {i + 1}/{len(manifest)}")

    result = pd.DataFrame(output_rows)
    manifest_path = os.path.join(
        OUTPUT_ROOT, f"{split_name}_crops_128_manifest.csv"
    )
    result.to_csv(manifest_path, index=False)

    print(f"\n{split_name}: created {len(result)} crops")
    print("Manifest:", manifest_path)
    return result

train_crops = create_crops(train_manifest, "train")
val_crops = create_crops(val_manifest, "val")

print("\nFinal crop counts:")
print("Train:", len(train_crops))
print("Validation:", len(val_crops))
```

```python

# ============================================================
# GEOVISION: ONE-CELL KAGGLE TRAINING PIPELINE
# ResNet18 | Disaster-wise split | Evaluation | Saved outputs
# ============================================================

import os, re, gc, json, time, random, shutil, warnings
from pathlib import Path

import numpy as np
import pandas as pd
from PIL import Image, ImageFile
ImageFile.LOAD_TRUNCATED_IMAGES = True

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")

# ---------------- CONFIG ----------------
SEED = 42
BATCH_SIZE = 32
EPOCHS = 10
PATIENCE = 3
IMAGE_SIZE = 224
NUM_WORKERS = 2

INPUT_ROOT = Path("/kaggle/input")
WORK_ROOT = Path("/kaggle/working")
OUT = WORK_ROOT / "GeoVision_Results"
CROP_ROOT = WORK_ROOT / "geovision_large_pilot" / "crops_128"

OUT.mkdir(parents=True, exist_ok=True)

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("=" * 65)
print("GEOVISION TRAINING PIPELINE")
print("Device:", DEVICE)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

print("Output:", OUT)
print("=" * 65)


# ---------------- FILE DISCOVERY ----------------
print("\n[1/8] Searching Kaggle input files...")

all_csv = list(INPUT_ROOT.rglob("*.csv"))
all_zip = list(INPUT_ROOT.rglob("*.zip"))

def find_csv(exact_name):
    matches = [
        p for p in all_csv
        if p.name.lower() == exact_name.lower()
    ]
    matches.sort(key=lambda p: (
        "geovision_large_pilot" not in str(p).lower(),
        len(str(p))
    ))
    return matches[0] if matches else None

train_manifest_path = find_csv("train_manifest.csv")
val_manifest_path = find_csv("val_manifest.csv")
train_crop_csv = find_csv("train_crops_128_manifest.csv")
val_crop_csv = find_csv("val_crops_128_manifest.csv")

print("Train manifest:", train_manifest_path)
print("Validation manifest:", val_manifest_path)
print("Train crop CSV:", train_crop_csv)
print("Validation crop CSV:", val_crop_csv)
print("ZIP files:", len(all_zip))

source_candidates = [
    INPUT_ROOT / "datasets/tunguz/"
    "xview2-challenge-dataset-train-and-test/"
    "train/train/images"
]

source_root = next(
    (p for p in source_candidates if p.is_dir()), None
)

if source_root is None:
    candidates = [
        p for p in INPUT_ROOT.rglob("images")
        if p.is_dir()
    ]
    candidates.sort(key=lambda p: (
        "xview2-challenge-dataset" not in str(p).lower(),
        len(str(p))
    ))
    source_root = candidates[0] if candidates else None

print("Source images:", source_root)


# ---------------- LOAD MANIFESTS ----------------
print("\n[2/8] Loading crop manifests...")

def load_crop_csv(path):
    if path is None:
        return None

    df = pd.read_csv(path)

    required = {"damage_label", "image_name", "building_id"}
    if not required.issubset(df.columns):
        print("Unexpected CSV columns:", path, df.columns.tolist())
        return None

    return df

train_df = load_crop_csv(train_crop_csv)
val_df = load_crop_csv(val_crop_csv)


# ---------------- OPTIONAL ZIP EXTRACTION ----------------
# Only extract ZIP files that look like GeoVision crop archives.
def extract_crop_archives():
    extracted = []

    for zpath in all_zip:
        name = zpath.name.lower()

        if not any(k in name for k in ["crop", "geovision", "pilot"]):
            continue

        dest = WORK_ROOT / ("extracted_" + zpath.stem)
        dest.mkdir(parents=True, exist_ok=True)

        try:
            with __import__("zipfile").ZipFile(zpath) as zf:
                members = zf.namelist()
                relevant = [
                    m for m in members
                    if m.lower().endswith((".png", ".jpg", ".jpeg", ".csv"))
                ]

                if not relevant:
                    continue

                # Avoid extracting huge unrelated archives.
                zf.extractall(dest)
                extracted.append(dest)
                print("Extracted:", zpath.name, "->", dest)

        except Exception as e:
            print("Could not extract", zpath.name, ":", str(e)[:150])

    return extracted

if train_df is None or val_df is None:
    extracted_roots = extract_crop_archives()

    # Search extracted CSV manifests.
    if extracted_roots:
        for root in extracted_roots:
            for p in root.rglob("*.csv"):
                if p.name.lower() == "train_crops_128_manifest.csv":
                    train_df = load_crop_csv(p)
                elif p.name.lower() == "val_crops_128_manifest.csv":
                    val_df = load_crop_csv(p)

                elif p.name.lower() == "train_manifest.csv" and train_manifest_path is None:
                    train_manifest_path = p
                elif p.name.lower() == "val_manifest.csv" and val_manifest_path is None:
                    val_manifest_path = p


# ---------------- REGENERATE CROPS IF NEEDED ----------------
def resolve_existing_path(value):
    p = Path(str(value))

    if p.is_file():
        return str(p)

    # Handle paths left over from previous Kaggle sessions.
    name = p.name

    for root in [INPUT_ROOT, WORK_ROOT]:
        for candidate in root.rglob(name):
            if candidate.is_file():
                return str(candidate)

    return str(p)


def generate_crops(manifest, split_name):
    rows = []
    root = CROP_ROOT / split_name
    root.mkdir(parents=True, exist_ok=True)

    for n, (_, row) in enumerate(manifest.iterrows(), 1):
        src = source_root / str(row["image_name"])

        if not src.is_file():
            continue

        try:
            with Image.open(src) as im:
                im = im.convert("RGB")
                width, height = im.size

                x1 = max(0, min(width, int(float(row["min_x"]))))
                y1 = max(0, min(height, int(float(row["min_y"]))))
                x2 = max(0, min(width, int(float(row["max_x"]) + 0.999)))
                y2 = max(0, min(height, int(float(row["max_y"]) + 0.999)))

                if x2 <= x1 or y2 <= y1:
                    continue

                crop = im.crop((x1, y1, x2, y2))
                crop = crop.resize((128, 128))

                label = str(row["damage_label"]).lower().strip()
                label_dir = root / label
                label_dir.mkdir(parents=True, exist_ok=True)

                filename = (
                    f"{Path(str(row['image_name'])).stem}_"
                    f"{row['building_id']}.png"
                )

                dest = label_dir / filename
                crop.save(dest)

                rows.append({
                    "resized_path": str(dest),
                    "damage_label": label,
                    "image_name": str(row["image_name"]),
                    "building_id": str(row["building_id"])
                })

        except Exception as e:
            print("Skipping image:", src.name, str(e)[:100])

        if n % 500 == 0:
            print(f"{split_name}: {n}/{len(manifest)}")

    return pd.DataFrame(rows)


def discover_crops():
    rows = []

    for root in [INPUT_ROOT, WORK_ROOT]:
        for p in root.rglob("*"):
            if not p.is_file():
                continue

            if p.suffix.lower() not in {".png", ".jpg", ".jpeg"}:
                continue

            parts = [x.lower() for x in p.parts]
            label = p.parent.name.lower()

            if label not in {
                "destroyed", "major-damage",
                "minor-damage", "no-damage"
            }:
                continue

            if "crops" not in parts and "crop" not in p.name.lower():
                continue

            # Best-effort recovery; preserve filename for disaster detection.
            stem = p.stem
            rows.append({
                "resized_path": str(p),
                "damage_label": label,
                "image_name": stem,
                "building_id": stem
            })

    return pd.DataFrame(rows)


if train_df is None or val_df is None:
    print("Crop manifests unavailable; searching for crop images...")
    found = discover_crops()

    if len(found):
        train_df = found.copy()
        val_df = found.copy()

    elif (
        train_manifest_path is not None
        and val_manifest_path is not None
        and source_root is not None
    ):
        print("Generating crops from original bounding-box manifests...")
        train_original = pd.read_csv(train_manifest_path)
        val_original = pd.read_csv(val_manifest_path)

        required = {
            "image_name", "building_id", "damage_label",
            "min_x", "min_y", "max_x", "max_y"
        }

        if not required.issubset(train_original.columns):
            raise ValueError("Train manifest is missing bounding-box columns.")

        if not required.issubset(val_original.columns):
            raise ValueError("Validation manifest is missing bounding-box columns.")

        train_df = generate_crops(train_original, "train")
        val_df = generate_crops(val_original, "val")

    else:
        raise FileNotFoundError(
            "Could not locate usable crop CSVs, crop images, or original "
            "manifests plus source images. Attach the correct dataset/ZIP "
            "to Kaggle Input and rerun."
        )


def normalize_manifest(df):
    df = df.copy()

    if "resized_path" not in df.columns:
        for col in ["crop_path", "image_path", "path"]:
            if col in df.columns:
                df = df.rename(columns={col: "resized_path"})
                break

    if "resized_path" not in df.columns:
        raise ValueError("Crop CSV needs a resized_path or image_path column.")

    df["resized_path"] = df["resized_path"].map(resolve_existing_path)
    df["damage_label"] = (
        df["damage_label"].astype(str).str.lower().str.strip()
    )

    return df

train_df = normalize_manifest(train_df)
val_df = normalize_manifest(val_df)

all_df = pd.concat([train_df, val_df], ignore_index=True)
all_df = all_df.drop_duplicates(subset=["resized_path"]).copy()

all_df = all_df[
    all_df["resized_path"].map(lambda x: Path(x).is_file())
].copy()

if all_df.empty:
    raise FileNotFoundError(
        "Crop manifest paths do not point to existing images. "
        "The ZIP may contain a different folder structure."
    )

print("Usable crops:", len(all_df))


# ---------------- DISASTER-WISE SPLIT ----------------
print("\n[3/8] Creating disaster-wise split...")

CLASSES = [
    "destroyed",
    "major-damage",
    "minor-damage",
    "no-damage"
]

all_df["damage_label"] = all_df["damage_label"].replace({
    "no_damage": "no-damage",
    "minor_damage": "minor-damage",
    "major_damage": "major-damage"
})

all_df = all_df[
    all_df["damage_label"].isin(CLASSES)
].copy()

# Recover disaster identity from original image names.
# Use conservative matching; do not silently invent a held-out split.
KNOWN_DISASTERS = [
    "hurricane-matthew",
    "hurricane-michael",
    "hurricane-florence",
    "hurricane-harvey",
    "guatemala-volcano",
    "mexico-earthquake",
    "midwest-flooding",
    "palu-tsunami",
    "santa-rosa-wildfire",
    "socal-fire",
    "nepal-flooding"
]

def identify_disaster(row):
    if "disaster" in row.index and pd.notna(row["disaster"]):
        value = str(row["disaster"]).lower().strip()
        if value and value != "nan":
            return value

    name = str(row.get("image_name", "")).lower()
    path = str(row.get("resized_path", "")).lower()
    text = name + " " + path

    for disaster in KNOWN_DISASTERS:
        if disaster in text:
            return disaster

    return "unknown"

all_df["disaster"] = all_df.apply(identify_disaster, axis=1)

# Do not make a fake held-out test set if disaster identity is unknown.
unknown_ratio = (all_df["disaster"] == "unknown").mean()

if unknown_ratio > 0.05:
    raise ValueError(
        f"{unknown_ratio:.1%} of crops have unknown disaster identity. "
        "The image_name/path values may have lost their original disaster "
        "names. Restore the original crop manifest or ZIP preserving names "
        "before running a valid unseen-disaster evaluation."
    )

held_out = {"hurricane-matthew", "hurricane-michael"}

test_df = all_df[
    all_df["disaster"].isin(held_out)
].copy()

trainval_df = all_df[
    ~all_df["disaster"].isin(held_out)
].copy()

if test_df.empty:
    raise ValueError(
        "No Hurricane Matthew/Michael records detected. "
        "Cannot run the planned held-out disaster test."
    )

disasters = sorted(trainval_df["disaster"].unique().tolist())

if len(disasters) < 2:
    raise ValueError(
        "At least two non-held-out disasters are needed for a "
        "disaster-wise train/validation split."
    )

random.Random(SEED).shuffle(disasters)
n_val_disasters = max(1, round(len(disasters) * 0.20))

val_disasters = set(disasters[:n_val_disasters])
train_disasters = set(disasters) - val_disasters

train_split = trainval_df[
    trainval_df["disaster"].isin(train_disasters)
].copy()

val_split = trainval_df[
    trainval_df["disaster"].isin(val_disasters)
].copy()

assert set(train_split.disaster).isdisjoint(set(val_split.disaster))
assert set(train_split.disaster).isdisjoint(held_out)
assert set(val_split.disaster).isdisjoint(held_out)

train_split.to_csv(OUT / "train_split.csv", index=False)
val_split.to_csv(OUT / "validation_split.csv", index=False)
test_df.to_csv(OUT / "test_split.csv", index=False)

print("Train samples:", len(train_split))
print("Train disasters:", sorted(train_split.disaster.unique()))
print("Validation samples:", len(val_split))
print("Validation disasters:", sorted(val_split.disaster.unique()))
print("Test samples:", len(test_df))
print("Test disasters:", sorted(test_df.disaster.unique()))


# ---------------- DATASETS ----------------
print("\n[4/8] Preparing data loaders...")

train_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomHorizontalFlip(0.5),
    transforms.RandomVerticalFlip(0.5),
    transforms.RandomRotation(15),
    transforms.ColorJitter(
        brightness=0.15, contrast=0.15, saturation=0.10
    ),
    transforms.ToTensor(),
    transforms.Normalize(
        [0.485, 0.456, 0.406],
        [0.229, 0.224, 0.225]
    )
])

eval_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        [0.485, 0.456, 0.406],
        [0.229, 0.224, 0.225]
    )
])


class GeoVisionDataset(Dataset):
    def __init__(self, df, transform):
        self.df = df.reset_index(drop=True)
        self.transform = transform
        self.label_map = {c: i for i, c in enumerate(CLASSES)}

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]

        with Image.open(row["resized_path"]) as im:
            image = self.transform(im.convert("RGB"))

        label = self.label_map[row["damage_label"]]
        return image, label


train_ds = GeoVisionDataset(train_split, train_transform)
val_ds = GeoVisionDataset(val_split, eval_transform)
test_ds = GeoVisionDataset(test_df, eval_transform)

loader_options = {
    "batch_size": BATCH_SIZE,
    "num_workers": NUM_WORKERS,
    "pin_memory": DEVICE.type == "cuda"
}

train_loader = DataLoader(
    train_ds, shuffle=True, **loader_options
)
val_loader = DataLoader(
    val_ds, shuffle=False, **loader_options
)
test_loader = DataLoader(
    test_ds, shuffle=False, **loader_options
)


# ---------------- MODEL ----------------
print("\n[5/8] Loading ResNet18...")

try:
    model = models.resnet18(
        weights=models.ResNet18_Weights.DEFAULT
    )
except Exception as e:
    print("Pretrained weights unavailable:", str(e)[:150])
    print("Using randomly initialized ResNet18.")
    model = models.resnet18(weights=None)

for param in model.parameters():
    param.requires_grad = False

for param in model.layer3.parameters():
    param.requires_grad = True

for param in model.layer4.parameters():
    param.requires_grad = True

model.fc = nn.Linear(model.fc.in_features, len(CLASSES))
model = model.to(DEVICE)

class_counts = train_split["damage_label"].value_counts()

weights = [
    len(train_split) / (len(CLASSES) * max(1, int(class_counts.get(c, 0))))
    for c in CLASSES
]

class_weights = torch.tensor(
    weights, dtype=torch.float32, device=DEVICE
)

criterion = nn.CrossEntropyLoss(weight=class_weights)

optimizer = torch.optim.AdamW(
    filter(lambda p: p.requires_grad, model.parameters()),
    lr=1e-4,
    weight_decay=1e-4
)

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode="max", factor=0.5, patience=1
)

use_amp = DEVICE.type == "cuda"
scaler = torch.amp.GradScaler("cuda", enabled=use_amp)


# ---------------- TRAINING ----------------
print("\n[6/8] Starting training...")

history = []
best_macro_f1 = -1.0
best_epoch = 0
bad_epochs = 0
best_path = OUT / "geovision_resnet18_best.pth"


def run_epoch(loader, training=False):
    model.train(training)

    total_loss = []
    true_labels = []
    predicted_labels = []

    for images, labels in loader:
        images = images.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)

        if training:
            optimizer.zero_grad(set_to_none=True)

        with torch.set_grad_enabled(training):
            with torch.autocast(
                device_type=DEVICE.type,
                enabled=use_amp
            ):
                logits = model(images)
                loss = criterion(logits, labels)

            if training:
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()

        total_loss.append(float(loss.detach().item()))
        predicted_labels.extend(
            logits.argmax(1).detach().cpu().tolist()
        )
        true_labels.extend(labels.detach().cpu().tolist())

    avg_loss = float(np.mean(total_loss)) if total_loss else 0.0
    accuracy = accuracy_score(true_labels, predicted_labels)
    macro_f1 = f1_score(
        true_labels, predicted_labels,
        average="macro", zero_division=0
    )

    return avg_loss, accuracy, macro_f1


start_time = time.time()

for epoch in range(1, EPOCHS + 1):
    train_loss, train_acc, train_f1 = run_epoch(
        train_loader, training=True
    )

    val_loss, val_acc, val_f1 = run_epoch(
        val_loader, training=False
    )

    scheduler.step(val_f1)

    row = {
        "epoch": epoch,
        "train_loss": train_loss,
        "train_accuracy": train_acc,
        "train_macro_f1": train_f1,
        "val_loss": val_loss,
        "val_accuracy": val_acc,
        "val_macro_f1": val_f1,
        "learning_rate": optimizer.param_groups[0]["lr"]
    }

    history.append(row)
    pd.DataFrame(history).to_csv(
        OUT / "training_history.csv", index=False
    )

    print(
        f"Epoch {epoch}/{EPOCHS} | "
        f"Train loss={train_loss:.4f}, acc={train_acc:.4f}, "
        f"F1={train_f1:.4f} | "
        f"Val loss={val_loss:.4f}, acc={val_acc:.4f}, "
        f"F1={val_f1:.4f}"
    )

    if val_f1 > best_macro_f1:
        best_macro_f1 = val_f1
        best_epoch = epoch
        bad_epochs = 0

        torch.save({
            "model_state_dict": model.state_dict(),
            "classes": CLASSES,
            "class_to_idx": {
                c: i for i, c in enumerate(CLASSES)
            },
            "image_size": IMAGE_SIZE,
            "best_epoch": best_epoch,
            "best_val_macro_f1": best_macro_f1,
            "seed": SEED
        }, best_path)

        print("  Best checkpoint saved.")

    else:
        bad_epochs += 1

        if bad_epochs >= PATIENCE:
            print("Early stopping.")
            break

print(f"Training duration: {(time.time() - start_time) / 60:.1f} minutes")


# ---------------- EVALUATION ----------------
print("\n[7/8] Evaluating best checkpoint...")

checkpoint = torch.load(
    best_path, map_location=DEVICE, weights_only=False
)
model.load_state_dict(checkpoint["model_state_dict"])


def predict(loader):
    model.eval()

    y_true = []
    y_pred = []

    with torch.no_grad():
        for images, labels in loader:
            images = images.to(DEVICE, non_blocking=True)

            with torch.autocast(
                device_type=DEVICE.type,
                enabled=use_amp
            ):
                logits = model(images)

            y_pred.extend(logits.argmax(1).cpu().tolist())
            y_true.extend(labels.tolist())

    return np.asarray(y_true), np.asarray(y_pred)


results = {}

for split_name, loader in [
    ("validation", val_loader),
    ("held_out_test", test_loader)
]:
    y_true, y_pred = predict(loader)

    # Corrected classification_report call.
    report = classification_report(
        y_true,
        y_pred,
        labels=list(range(len(CLASSES))),
        target_names=CLASSES,
        output_dict=True,
        zero_division=0
    )

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=list(range(len(CLASSES)))
    )

    metrics = {
        "accuracy": float(
            accuracy_score(y_true, y_pred)
        ),
        "macro_f1": float(
            f1_score(
                y_true, y_pred,
                average="macro", zero_division=0
            )
        ),
        "weighted_f1": float(
            f1_score(
                y_true, y_pred,
                average="weighted", zero_division=0
            )
        ),
        "n_samples": int(len(y_true)),
        "classification_report": report,
        "confusion_matrix": cm.tolist()
    }

    results[split_name] = metrics

    pd.DataFrame(report).transpose().to_csv(
        OUT / f"{split_name}_classification_report.csv"
    )

    pd.DataFrame(
        cm, index=CLASSES, columns=CLASSES
    ).to_csv(OUT / f"{split_name}_confusion_matrix.csv")

    plt.figure(figsize=(8, 6))
    plt.imshow(cm, interpolation="nearest")
    plt.title(f"GeoVision — {split_name}")
    plt.colorbar()
    plt.xticks(
        range(len(CLASSES)), CLASSES,
        rotation=35, ha="right"
    )
    plt.yticks(range(len(CLASSES)), CLASSES)
    plt.xlabel("Predicted label")
    plt.ylabel("True label")

    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            plt.text(
                j, i, str(cm[i, j]),
                ha="center", va="center"
            )

    plt.tight_layout()
    plt.savefig(
        OUT / f"{split_name}_confusion_matrix.png",
        dpi=160
    )
    plt.show()
    plt.close()

    print(f"\n{split_name.upper()} RESULTS")
    print("Accuracy:", round(metrics["accuracy"], 4))
    print("Macro F1:", round(metrics["macro_f1"], 4))
    print("Weighted F1:", round(metrics["weighted_f1"], 4))

    print(
        classification_report(
            y_true,
            y_pred,
            labels=list(range(len(CLASSES))),
            target_names=CLASSES,
            zero_division=0
        )
    )


# ---------------- SAVE PLOTS / RESULTS ----------------
history_df = pd.DataFrame(history)
history_df.to_csv(OUT / "training_history.csv", index=False)

if len(history_df):
    plt.figure(figsize=(8, 5))
    plt.plot(
        history_df["epoch"], history_df["train_loss"],
        label="Train loss"
    )
    plt.plot(
        history_df["epoch"], history_df["val_loss"],
        label="Validation loss"
    )
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title("GeoVision Training Loss")
    plt.legend()
    plt.tight_layout()
    plt.savefig(OUT / "training_loss.png", dpi=160)
    plt.show()
    plt.close()

    plt.figure(figsize=(8, 5))
    plt.plot(
        history_df["epoch"], history_df["train_macro_f1"],
        label="Train macro-F1"
    )
    plt.plot(
        history_df["epoch"], history_df["val_macro_f1"],
        label="Validation macro-F1"
    )
    plt.xlabel("Epoch")
    plt.ylabel("Macro-F1")
    plt.title("GeoVision Macro-F1")
    plt.legend()
    plt.tight_layout()
    plt.savefig(OUT / "training_macro_f1.png", dpi=160)
    plt.show()
    plt.close()


summary = {
    "project": "GeoVision",
    "model": "ResNet18",
    "device": str(DEVICE),
    "classes": CLASSES,
    "train_samples": int(len(train_split)),
    "validation_samples": int(len(val_split)),
    "test_samples": int(len(test_df)),
    "train_disasters": sorted(
        train_split["disaster"].unique().tolist()
    ),
    "validation_disasters": sorted(
        val_split["disaster"].unique().tolist()
    ),
    "test_disasters": sorted(
        test_df["disaster"].unique().tolist()
    ),
    "best_epoch": int(best_epoch),
    "best_validation_macro_f1": float(best_macro_f1),
    "metrics": results
}

with open(OUT / "evaluation_metrics.json", "w") as f:
    json.dump(results, f, indent=2)

with open(OUT / "run_summary.json", "w") as f:
    json.dump(summary, f, indent=2)

with open(OUT / "environment.txt", "w") as f:
    f.write(f"Python: {os.sys.version}\n")
    f.write(f"PyTorch: {torch.__version__}\n")
    f.write(f"CUDA available: {torch.cuda.is_available()}\n")
    f.write(f"Device: {DEVICE}\n")


# ---------------- ZIP OUTPUTS ----------------
print("\n[8/8] Packaging results...")

zip_path = shutil.make_archive(
    str(WORK_ROOT / "GeoVision_Results"),
    "zip",
    OUT
)

print("\n" + "=" * 65)
print("GEOVISION PIPELINE FINISHED")
print("Best model:", best_path)
print("Results folder:", OUT)
print("Results ZIP:", zip_path)
print("\nOutput files:")

for p in sorted(OUT.iterdir()):
    if p.is_file():
        size_mb = p.stat().st_size / (1024 * 1024)
        print(f"{p.name} ({size_mb:.2f} MB)")

print("=" * 65)
print("Download GeoVision_Results.zip from the Kaggle Output panel.")
print("For long-term persistence, download it or save it as a Kaggle Dataset.")
```

```python

# ================================================================
# GEOVISION V2 — DISASTER-LEVEL DAMAGE CLASSIFICATION
# Single-cell Kaggle pipeline | ResNet18 | GPU | Class balancing
# ================================================================

import os, json, random, shutil, zipfile, time, copy, warnings
from pathlib import Path
from collections import Counter

import numpy as np
import pandas as pd
from PIL import Image, ImageFile
ImageFile.LOAD_TRUNCATED_IMAGES = True

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from torchvision import transforms, models

from sklearn.metrics import (
    accuracy_score, f1_score, classification_report,
    confusion_matrix, precision_recall_fscore_support
)
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")

# -------------------- 1. SETTINGS --------------------

SEED = 42
BATCH_SIZE = 32
EPOCHS = 12
PATIENCE = 4
IMAGE_SIZE = 128
NUM_WORKERS = 2
LR_HEAD = 3e-4
LR_BACKBONE = 3e-5

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)
if DEVICE.type == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))

# Locate the previous run's split CSVs.
OLD_DIR = Path("/kaggle/working/GeoVision_Results")
SPLIT_FILES = {
    "train": OLD_DIR / "train_split.csv",
    "val": OLD_DIR / "validation_split.csv",
    "test": OLD_DIR / "test_split.csv"
}

for name, path in SPLIT_FILES.items():
    if not path.exists():
        raise FileNotFoundError(
            f"Cannot find {path}\n"
            "Make sure the previous GeoVision results are available "
            "in /kaggle/working/GeoVision_Results. If Kaggle restarted, "
            "add your previous results ZIP as a Kaggle input and update "
            "OLD_DIR to the extracted folder."
        )

# New output directory; do not overwrite the previous results.
OUT = Path("/kaggle/working/GeoVision_V2_Results")
OUT.mkdir(parents=True, exist_ok=True)

# -------------------- 2. LOAD & CORRECT SPLITS --------------------

train_old = pd.read_csv(SPLIT_FILES["train"])
val_old   = pd.read_csv(SPLIT_FILES["val"])
test_df   = pd.read_csv(SPLIT_FILES["test"])

required = {"resized_path", "damage_label", "disaster"}
for name, df in [
    ("train", train_old), ("validation", val_old), ("test", test_df)
]:
    missing = required - set(df.columns)
    if missing:
        raise ValueError(f"{name} CSV is missing columns: {missing}")

# Keep Matthew/Michael completely untouched as the held-out test set.
test_disasters = set(test_df["disaster"].dropna().unique())
expected_test = {"hurricane-matthew", "hurricane-michael"}
if not expected_test.issubset(test_disasters):
    raise ValueError(
        f"Expected Matthew and Michael in test set, found: {test_disasters}"
    )

# Validation will use Florence + Midwest flooding.
# Mexico earthquake (previous validation) moves into training.
# This avoids mixing disasters across train and validation.
florence = train_old[
    train_old["disaster"].astype(str).eq("hurricane-florence")
].copy()

midwest = val_old[
    val_old["disaster"].astype(str).eq("midwest-flooding")
].copy()

mexico = val_old[
    val_old["disaster"].astype(str).eq("mexico-earthquake")
].copy()

train_df = pd.concat([
    train_old[
        ~train_old["disaster"].astype(str).eq("hurricane-florence")
    ],
    mexico
], ignore_index=True)

val_df = pd.concat([florence, midwest], ignore_index=True)

# Check for accidental disaster overlap.
train_dis = set(train_df["disaster"].dropna().unique())
val_dis = set(val_df["disaster"].dropna().unique())

assert train_dis.isdisjoint(val_dis), (
    f"Train/validation disaster overlap: {train_dis & val_dis}"
)
assert train_dis.isdisjoint(test_disasters), (
    f"Train/test disaster overlap: {train_dis & test_disasters}"
)
assert val_dis.isdisjoint(test_disasters), (
    f"Validation/test disaster overlap: {val_dis & test_disasters}"
)

# Check image paths before training.
for split_name, df in [
    ("train", train_df), ("validation", val_df), ("test", test_df)
]:
    exists = df["resized_path"].astype(str).map(os.path.isfile)
    if not exists.all():
        missing_paths = df.loc[~exists, "resized_path"].head(5).tolist()
        raise FileNotFoundError(
            f"{split_name}: {int((~exists).sum())} image paths are missing.\n"
            f"Examples: {missing_paths}"
        )

# Four consistent classes.
CLASSES = ["destroyed", "major-damage", "minor-damage", "no-damage"]
class_to_idx = {name: i for i, name in enumerate(CLASSES)}

for split_name, df in [
    ("train", train_df), ("validation", val_df), ("test", test_df)
]:
    unknown = set(df["damage_label"].dropna().unique()) - set(CLASSES)
    if unknown:
        raise ValueError(f"Unknown labels in {split_name}: {unknown}")
    if df["damage_label"].isna().any():
        raise ValueError(f"Missing labels in {split_name}")

print("\n" + "=" * 65)
print("GEOVISION V2 — SPLIT SUMMARY")
print("=" * 65)

for name, df in [
    ("TRAIN", train_df), ("VALIDATION", val_df), ("TEST", test_df)
]:
    print(f"\n{name}: {len(df)} images")
    print("Disasters:", sorted(df["disaster"].dropna().unique()))
    print(df["damage_label"].value_counts().reindex(CLASSES, fill_value=0))
    print("Class percentages:")
    print(
        (df["damage_label"].value_counts(normalize=True)
         .reindex(CLASSES, fill_value=0) * 100).round(1)
    )

# Save exact split definitions for reproducibility.
train_df.to_csv(OUT / "train_split.csv", index=False)
val_df.to_csv(OUT / "validation_split.csv", index=False)
test_df.to_csv(OUT / "test_split.csv", index=False)

# -------------------- 3. DATA PIPELINE --------------------

train_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomVerticalFlip(),
    transforms.RandomRotation(12),
    transforms.ColorJitter(brightness=0.12, contrast=0.12, saturation=0.08),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

eval_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

class DamageDataset(Dataset):
    def __init__(self, frame, transform):
        self.frame = frame.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.frame)

    def __getitem__(self, index):
        row = self.frame.iloc[index]
        image = Image.open(row["resized_path"]).convert("RGB")
        image = self.transform(image)
        label = class_to_idx[row["damage_label"]]
        return image, label

train_dataset = DamageDataset(train_df, train_transform)
val_dataset = DamageDataset(val_df, eval_transform)
test_dataset = DamageDataset(test_df, eval_transform)

# Use inverse-square-root frequency sampling to reduce imbalance
# without oversampling rare classes as aggressively as inverse frequency.
train_labels = np.array([
    class_to_idx[label] for label in train_df["damage_label"]
])
counts = np.bincount(train_labels, minlength=len(CLASSES))
sample_class_weights = 1.0 / np.sqrt(np.maximum(counts, 1))
sample_weights = sample_class_weights[train_labels]

sampler = WeightedRandomSampler(
    weights=torch.as_tensor(sample_weights, dtype=torch.double),
    num_samples=len(train_dataset),
    replacement=True
)

train_loader = DataLoader(
    train_dataset, batch_size=BATCH_SIZE, sampler=sampler,
    num_workers=NUM_WORKERS, pin_memory=(DEVICE.type == "cuda")
)
val_loader = DataLoader(
    val_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=(DEVICE.type == "cuda")
)
test_loader = DataLoader(
    test_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=(DEVICE.type == "cuda")
)

# -------------------- 4. MODEL --------------------

print("\nLoading pretrained ResNet18...")
try:
    weights = models.ResNet18_Weights.DEFAULT
    model = models.resnet18(weights=weights)
    print("Pretrained ImageNet weights loaded.")
except Exception as e:
    print("WARNING: pretrained weights unavailable:", repr(e))
    print("Using randomly initialized ResNet18 instead.")
    model = models.resnet18(weights=None)

# Freeze early layers; fine-tune layer3, layer4 and classifier.
for p in model.parameters():
    p.requires_grad = False

for p in model.layer3.parameters():
    p.requires_grad = True
for p in model.layer4.parameters():
    p.requires_grad = True

in_features = model.fc.in_features
model.fc = nn.Sequential(
    nn.Dropout(0.35),
    nn.Linear(in_features, len(CLASSES))
)
for p in model.fc.parameters():
    p.requires_grad = True

model = model.to(DEVICE)

# Weighted loss as a second, moderate imbalance correction.
# Normalize weights to have mean approximately 1.
class_weights = 1.0 / np.sqrt(np.maximum(counts, 1))
class_weights = class_weights / class_weights.mean()
criterion = nn.CrossEntropyLoss(
    weight=torch.tensor(class_weights, dtype=torch.float32, device=DEVICE)
)

optimizer = torch.optim.AdamW([
    {"params": model.layer3.parameters(), "lr": LR_BACKBONE},
    {"params": model.layer4.parameters(), "lr": LR_BACKBONE},
    {"params": model.fc.parameters(), "lr": LR_HEAD}
], weight_decay=1e-4)

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode="max", factor=0.5, patience=1
)

# -------------------- 5. TRAINING --------------------

def run_epoch(loader, training):
    model.train() if training else model.eval()
    total_loss = 0.0
    total = 0
    y_true, y_pred = [], []

    context = torch.enable_grad() if training else torch.no_grad()
    with context:
        for images, labels in loader:
            images = images.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)

            if training:
                optimizer.zero_grad(set_to_none=True)

            logits = model(images)
            loss = criterion(logits, labels)

            if training:
                loss.backward()
                torch.nn.utils.clip_grad_norm_(
                    [p for p in model.parameters() if p.requires_grad], 2.0
                )
                optimizer.step()

            batch_size = labels.size(0)
            total_loss += loss.item() * batch_size
            total += batch_size
            preds = logits.argmax(dim=1)

            y_true.extend(labels.detach().cpu().numpy().tolist())
            y_pred.extend(preds.detach().cpu().numpy().tolist())

    return {
        "loss": total_loss / max(total, 1),
        "accuracy": accuracy_score(y_true, y_pred),
        "macro_f1": f1_score(
            y_true, y_pred, average="macro", zero_division=0
        ),
        "y_true": y_true,
        "y_pred": y_pred
    }

history = []
best_f1 = -1.0
best_epoch = 0
best_state = None
epochs_without_improvement = 0
start_time = time.time()

print("\n" + "=" * 65)
print("STARTING GEOVISION V2 TRAINING")
print("=" * 65)

for epoch in range(1, EPOCHS + 1):
    train_result = run_epoch(train_loader, training=True)
    val_result = run_epoch(val_loader, training=False)

    scheduler.step(val_result["macro_f1"])

    row = {
        "epoch": epoch,
        "train_loss": train_result["loss"],
        "train_accuracy": train_result["accuracy"],
        "train_macro_f1": train_result["macro_f1"],
        "val_loss": val_result["loss"],
        "val_accuracy": val_result["accuracy"],
        "val_macro_f1": val_result["macro_f1"]
    }
    history.append(row)

    print(
        f"Epoch {epoch:02d}/{EPOCHS} | "
        f"Train loss {row['train_loss']:.4f} | "
        f"Train acc {row['train_accuracy']:.4f} | "
        f"Val loss {row['val_loss']:.4f} | "
        f"Val acc {row['val_accuracy']:.4f} | "
        f"Val macro-F1 {row['val_macro_f1']:.4f}"
    )

    # Save best model by validation macro-F1, not test performance.
    if val_result["macro_f1"] > best_f1:
        best_f1 = val_result["macro_f1"]
        best_epoch = epoch
        best_state = copy.deepcopy(model.state_dict())
        torch.save({
            "epoch": best_epoch,
            "model_state_dict": best_state,
            "classes": CLASSES,
            "image_size": IMAGE_SIZE,
            "validation_macro_f1": best_f1
        }, OUT / "geovision_resnet18_v2_best.pth")
        epochs_without_improvement = 0
    else:
        epochs_without_improvement += 1

    if epochs_without_improvement >= PATIENCE:
        print(f"Early stopping at epoch {epoch}.")
        break

pd.DataFrame(history).to_csv(OUT / "training_history.csv", index=False)

if best_state is None:
    raise RuntimeError("Training did not produce a valid checkpoint.")

model.load_state_dict(best_state)
print(f"\nBest epoch: {best_epoch}; validation macro-F1: {best_f1:.4f}")

# -------------------- 6. FINAL HELD-OUT TEST EVALUATION --------------------
# Evaluate Matthew/Michael only after model selection is complete.

test_result = run_epoch(test_loader, training=False)
y_true = np.array(test_result["y_true"])
y_pred = np.array(test_result["y_pred"])

report_dict = classification_report(
    y_true, y_pred, labels=list(range(len(CLASSES))),
    target_names=CLASSES, output_dict=True, zero_division=0
)
report_text = classification_report(
    y_true, y_pred, labels=list(range(len(CLASSES))),
    target_names=CLASSES, zero_division=0
)

cm = confusion_matrix(y_true, y_pred, labels=list(range(len(CLASSES))))
pd.DataFrame(cm, index=CLASSES, columns=CLASSES).to_csv(
    OUT / "test_confusion_matrix.csv"
)
pd.DataFrame(report_dict).transpose().to_csv(
    OUT / "test_classification_report.csv"
)

# Validation report as well.
val_result = run_epoch(val_loader, training=False)
val_report = classification_report(
    val_result["y_true"], val_result["y_pred"],
    labels=list(range(len(CLASSES))), target_names=CLASSES,
    output_dict=True, zero_division=0
)
pd.DataFrame(val_report).transpose().to_csv(
    OUT / "validation_classification_report.csv"
)
pd.DataFrame(
    confusion_matrix(
        val_result["y_true"], val_result["y_pred"],
        labels=list(range(len(CLASSES)))
    ),
    index=CLASSES, columns=CLASSES
).to_csv(OUT / "validation_confusion_matrix.csv")

# -------------------- 7. PLOTS --------------------

hist_df = pd.DataFrame(history)

plt.figure(figsize=(9, 5))
plt.plot(hist_df["epoch"], hist_df["train_loss"], marker="o", label="Train loss")
plt.plot(hist_df["epoch"], hist_df["val_loss"], marker="o", label="Validation loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("GeoVision V2 — Training and Validation Loss")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(OUT / "loss_curve.png", dpi=160)
plt.close()

plt.figure(figsize=(9, 5))
plt.plot(hist_df["epoch"], hist_df["train_macro_f1"], marker="o", label="Train macro-F1")
plt.plot(hist_df["epoch"], hist_df["val_macro_f1"], marker="o", label="Validation macro-F1")
plt.xlabel("Epoch")
plt.ylabel("Macro-F1")
plt.title("GeoVision V2 — Macro-F1")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(OUT / "macro_f1_curve.png", dpi=160)
plt.close()

plt.figure(figsize=(7, 6))
plt.imshow(cm, interpolation="nearest")
plt.title("GeoVision V2 — Held-out Test Confusion Matrix")
plt.colorbar()
ticks = np.arange(len(CLASSES))
plt.xticks(ticks, CLASSES, rotation=35, ha="right")
plt.yticks(ticks, CLASSES)
plt.xlabel("Predicted label")
plt.ylabel("True label")
threshold = cm.max() / 2 if cm.max() else 0
for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(
            j, i, str(cm[i, j]), ha="center", va="center",
            color="white" if cm[i, j] > threshold else "black"
        )
plt.tight_layout()
plt.savefig(OUT / "test_confusion_matrix.png", dpi=160)
plt.close()

# -------------------- 8. SAVE SUMMARY --------------------

elapsed_minutes = (time.time() - start_time) / 60

summary = {
    "project": "GeoVision",
    "version": "V2",
    "model": "ResNet18",
    "device": str(DEVICE),
    "classes": CLASSES,
    "image_size": IMAGE_SIZE,
    "best_epoch": int(best_epoch),
    "best_validation_macro_f1": float(best_f1),
    "validation_accuracy": float(val_result["accuracy"]),
    "validation_macro_f1": float(val_result["macro_f1"]),
    "test_accuracy": float(test_result["accuracy"]),
    "test_macro_f1": float(test_result["macro_f1"]),
    "test_weighted_f1": float(f1_score(
        y_true, y_pred, average="weighted", zero_division=0
    )),
    "test_disasters": sorted(test_disasters),
    "train_disasters": sorted(train_dis),
    "validation_disasters": sorted(val_dis),
    "train_samples": int(len(train_df)),
    "validation_samples": int(len(val_df)),
    "test_samples": int(len(test_df)),
    "training_minutes": round(elapsed_minutes, 2),
    "note": (
        "Matthew and Michael were held out from training and validation. "
        "Test metrics are for this held-out disaster split."
    )
}

with open(OUT / "evaluation_metrics.json", "w") as f:
    json.dump({
        "test_accuracy": summary["test_accuracy"],
        "test_macro_f1": summary["test_macro_f1"],
        "test_weighted_f1": summary["test_weighted_f1"],
        "validation_accuracy": summary["validation_accuracy"],
        "validation_macro_f1": summary["validation_macro_f1"],
        "classification_report": report_dict
    }, f, indent=2)

with open(OUT / "run_summary.json", "w") as f:
    json.dump(summary, f, indent=2)

with open(OUT / "environment.txt", "w") as f:
    f.write(f"Python: {os.sys.version}\n")
    f.write(f"PyTorch: {torch.__version__}\n")
    f.write(f"Device: {DEVICE}\n")
    if DEVICE.type == "cuda":
        f.write(f"GPU: {torch.cuda.get_device_name(0)}\n")

# -------------------- 9. ZIP RESULTS --------------------

zip_path = Path("/kaggle/working/GeoVision_V2_Results.zip")
if zip_path.exists():
    zip_path.unlink()

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as zf:
    for file in OUT.rglob("*"):
        if file.is_file():
            zf.write(file, arcname=f"GeoVision_V2_Results/{file.relative_to(OUT)}")

# -------------------- 10. FINAL REPORT --------------------

print("\n" + "=" * 65)
print("GEOVISION V2 — FINAL RESULTS")
print("=" * 65)
print(f"Best epoch:             {best_epoch}")
print(f"Validation accuracy:    {summary['validation_accuracy']:.4f}")
print(f"Validation macro-F1:    {summary['validation_macro_f1']:.4f}")
print(f"Test accuracy:          {summary['test_accuracy']:.4f}")
print(f"Test macro-F1:          {summary['test_macro_f1']:.4f}")
print(f"Test weighted F1:       {summary['test_weighted_f1']:.4f}")
print(f"Training time:          {elapsed_minutes:.2f} minutes")

print("\nHELD-OUT TEST CLASSIFICATION REPORT")
print(report_text)

print("\nFiles saved in:", OUT)
print("ZIP file:", zip_path)
print("ZIP size:", round(zip_path.stat().st_size / (1024 * 1024), 2), "MB")
print("\nTraining complete. Download GeoVision_V2_Results.zip from Kaggle Output.")
```

```python

import torch
import pandas as pd
import numpy as np
from pathlib import Path
from collections import Counter
from PIL import Image
from torchvision import models, transforms
from torch.utils.data import Dataset, DataLoader
import torch.nn as nn
from sklearn.metrics import confusion_matrix, classification_report

OUT = Path("/kaggle/working/GeoVision_V2_Results")
CSV = OUT / "test_split.csv"
CKPT = OUT / "geovision_resnet18_v2_best.pth"

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
classes = ["destroyed", "major-damage", "minor-damage", "no-damage"]
class_to_idx = {c: i for i, c in enumerate(classes)}

df = pd.read_csv(CSV)
transform = transforms.Compose([
    transforms.Resize((128, 128)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

class TestDataset(Dataset):
    def __init__(self, frame):
        self.frame = frame.reset_index(drop=True)
    def __len__(self):
        return len(self.frame)
    def __getitem__(self, i):
        row = self.frame.iloc[i]
        image = Image.open(row["resized_path"]).convert("RGB")
        return transform(image), class_to_idx[row["damage_label"]]

loader = DataLoader(TestDataset(df), batch_size=32, shuffle=False, num_workers=2)

model = models.resnet18(weights=None)
model.fc = nn.Sequential(nn.Dropout(0.35), nn.Linear(model.fc.in_features, 4))
checkpoint = torch.load(CKPT, map_location=device, weights_only=False)
model.load_state_dict(checkpoint["model_state_dict"])
model.to(device).eval()

y_true, y_pred = [], []
with torch.no_grad():
    for images, labels in loader:
        logits = model(images.to(device))
        predictions = logits.argmax(1).cpu().numpy()
        y_pred.extend(predictions.tolist())
        y_true.extend(labels.numpy().tolist())

print("Checkpoint epoch:", checkpoint["epoch"])
print("\nTRUE CLASS COUNTS")
print(pd.Series(y_true).map(dict(enumerate(classes))).value_counts())

print("\nPREDICTED CLASS COUNTS")
print(pd.Series(y_pred).map(dict(enumerate(classes))).value_counts())

print("\nCONFUSION MATRIX (rows=true, columns=predicted)")
print(pd.DataFrame(
    confusion_matrix(y_true, y_pred, labels=range(4)),
    index=classes, columns=classes
))

print("\nCLASSIFICATION REPORT")
print(classification_report(
    y_true, y_pred, labels=range(4),
    target_names=classes, zero_division=0
))
```

```python

import pandas as pd
import os
from pathlib import Path
from PIL import Image
from collections import defaultdict

csv_path = Path("/kaggle/working/GeoVision_V2_Results/train_split.csv")
df = pd.read_csv(csv_path)

print("=" * 55)
print("GEOVISION — IMAGE QUALITY DIAGNOSTICS")
print("=" * 55)

print("\n1. Image counts by class:")
print(df["damage_label"].value_counts().to_string())

print("\n2. Missing image files:")
exists = df["resized_path"].astype(str).map(os.path.isfile)
print("Missing:", int((~exists).sum()), "of", len(df))

print("\n3. Image dimensions and modes:")
dimension_counts = defaultdict(int)
mode_counts = defaultdict(int)
errors = []

for _, row in df.iterrows():
    path = row["resized_path"]
    try:
        with Image.open(path) as im:
            dimension_counts[str(im.size)] += 1
            mode_counts[im.mode] += 1
            im.verify()
    except Exception as e:
        errors.append((path, str(e)))

print("Dimensions:", dict(dimension_counts))
print("Image modes:", dict(mode_counts))
print("Unreadable/corrupt images:", len(errors))

if errors:
    print("Examples of errors:")
    for path, error in errors[:5]:
        print(path, "->", error[:150])

print("\n4. Class counts by disaster:")
print(pd.crosstab(df["disaster"], df["damage_label"]).to_string())

print("\n5. Duplicate path check:")
print("Duplicate paths:", int(df["resized_path"].duplicated().sum()))

if "image_name" in df.columns:
    print("Duplicate image names:", int(df["image_name"].duplicated().sum()))

print("\nDiagnostics complete.")
```

```python

import pandas as pd
import hashlib
from pathlib import Path
from collections import Counter, defaultdict

csv_path = Path("/kaggle/working/GeoVision_V2_Results/train_split.csv")
df = pd.read_csv(csv_path)

print("=" * 60)
print("GEOVISION — DUPLICATE IMAGE INVESTIGATION")
print("=" * 60)

# 1. Repeated filenames and disaster overlap
df["filename"] = df["resized_path"].astype(str).map(
    lambda p: Path(p).name
)

name_counts = df["filename"].value_counts()
repeated = name_counts[name_counts > 1]

print("\nRepeated filenames:", len(repeated))
print("Rows with repeated filenames:",
      int(df["filename"].isin(repeated.index).sum()))

if len(repeated):
    print("\nExamples of repeated filenames:")
    for filename in repeated.index[:10]:
        rows = df[df["filename"] == filename]
        print(
            filename,
            "| disasters:", sorted(rows["disaster"].astype(str).unique()),
            "| labels:", sorted(rows["damage_label"].astype(str).unique()),
            "| paths:", rows["resized_path"].nunique()
        )

# 2. Exact file-content hashes (first 500 repeated-name rows)
# Hashing detects byte-identical files even if their paths differ.
sample_paths = (
    df[df["filename"].isin(repeated.index)]["resized_path"]
    .drop_duplicates().head(500).tolist()
)

hash_to_paths = defaultdict(list)

for path in sample_paths:
    try:
        h = hashlib.sha256()
        with open(path, "rb") as f:
            for chunk in iter(lambda: f.read(1024 * 1024), b""):
                h.update(chunk)
        hash_to_paths[h.hexdigest()].append(path)
    except Exception as e:
        print("Could not hash:", path, str(e)[:100])

exact_duplicates = {
    h: paths for h, paths in hash_to_paths.items()
    if len(paths) > 1
}

print("\nUnique files hashed:", len(sample_paths))
print("Byte-identical duplicate groups in this sample:",
      len(exact_duplicates))

for paths in list(exact_duplicates.values())[:5]:
    print("Duplicate group:", paths)

print("\nInvestigation complete.")
```

```python

# ================================================================
# GEOVISION V3 — IMPROVED RESNET18 TRAINING
# Same V2 split | No weighted sampler | Moderate class-weighted loss
# ================================================================

import os, json, time, copy, random, zipfile, warnings
from pathlib import Path

import numpy as np
import pandas as pd
from PIL import Image, ImageFile
ImageFile.LOAD_TRUNCATED_IMAGES = True
warnings.filterwarnings("ignore")

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models

from sklearn.metrics import (
    accuracy_score, f1_score, classification_report, confusion_matrix
)
import matplotlib.pyplot as plt

# -------------------- 1. SETTINGS --------------------

SEED = 42
IMAGE_SIZE = 128
BATCH_SIZE = 32
EPOCHS = 15
PATIENCE = 4
NUM_WORKERS = 2

BACKBONE_LR = 1e-5
HEAD_LR = 1e-4

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)
if DEVICE.type == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))

V2_DIR = Path("/kaggle/working/GeoVision_V2_Results")
OUT = Path("/kaggle/working/GeoVision_V3_Results")
OUT.mkdir(parents=True, exist_ok=True)

for filename in [
    "train_split.csv", "validation_split.csv", "test_split.csv"
]:
    if not (V2_DIR / filename).is_file():
        raise FileNotFoundError(f"Required V2 file missing: {V2_DIR / filename}")

# -------------------- 2. LOAD SAME SPLITS --------------------

train_df = pd.read_csv(V2_DIR / "train_split.csv")
val_df = pd.read_csv(V2_DIR / "validation_split.csv")
test_df = pd.read_csv(V2_DIR / "test_split.csv")

CLASSES = ["destroyed", "major-damage", "minor-damage", "no-damage"]
CLASS_TO_IDX = {c: i for i, c in enumerate(CLASSES)}

required_cols = {"resized_path", "damage_label", "disaster"}

for split_name, df in [
    ("train", train_df), ("validation", val_df), ("test", test_df)
]:
    if not required_cols.issubset(df.columns):
        raise ValueError(f"{split_name} CSV has missing required columns.")
    if df["damage_label"].isna().any():
        raise ValueError(f"{split_name} contains missing labels.")
    if set(df["damage_label"].unique()) - set(CLASSES):
        raise ValueError(f"{split_name} has unexpected class labels.")
    missing = (~df["resized_path"].astype(str).map(os.path.isfile)).sum()
    if missing:
        raise FileNotFoundError(
            f"{split_name} has {missing} missing image paths."
        )

train_dis = set(train_df["disaster"].dropna().unique())
val_dis = set(val_df["disaster"].dropna().unique())
test_dis = set(test_df["disaster"].dropna().unique())

assert train_dis.isdisjoint(val_dis), "Train/validation disaster overlap!"
assert train_dis.isdisjoint(test_dis), "Train/test disaster overlap!"
assert val_dis.isdisjoint(test_dis), "Validation/test disaster overlap!"

assert {"hurricane-matthew", "hurricane-michael"}.issubset(test_dis), (
    "Expected Matthew and Michael in held-out test split."
)

print("\nSPLIT SUMMARY")
for name, df in [
    ("TRAIN", train_df), ("VALIDATION", val_df), ("TEST", test_df)
]:
    print(f"\n{name}: {len(df)} images")
    print("Disasters:", sorted(df["disaster"].dropna().unique()))
    print(df["damage_label"].value_counts().reindex(CLASSES, fill_value=0))

# Copy split definitions into the new results folder.
train_df.to_csv(OUT / "train_split.csv", index=False)
val_df.to_csv(OUT / "validation_split.csv", index=False)
test_df.to_csv(OUT / "test_split.csv", index=False)

# -------------------- 3. TRANSFORMS & DATASETS --------------------

train_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomVerticalFlip(),
    transforms.RandomRotation(10),
    transforms.ColorJitter(brightness=0.08, contrast=0.08),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

eval_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

class DamageDataset(Dataset):
    def __init__(self, frame, transform):
        self.frame = frame.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.frame)

    def __getitem__(self, index):
        row = self.frame.iloc[index]
        with Image.open(row["resized_path"]) as im:
            image = im.convert("RGB")
        image = self.transform(image)
        label = CLASS_TO_IDX[row["damage_label"]]
        return image, label

train_dataset = DamageDataset(train_df, train_transform)
val_dataset = DamageDataset(val_df, eval_transform)
test_dataset = DamageDataset(test_df, eval_transform)

train_loader = DataLoader(
    train_dataset, batch_size=BATCH_SIZE, shuffle=True,
    num_workers=NUM_WORKERS, pin_memory=(DEVICE.type == "cuda")
)
val_loader = DataLoader(
    val_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=(DEVICE.type == "cuda")
)
test_loader = DataLoader(
    test_dataset, batch_size=BATCH_SIZE, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=(DEVICE.type == "cuda")
)

# -------------------- 4. MODEL --------------------

print("\nLoading pretrained ResNet18...")
try:
    model = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)
    print("Pretrained ImageNet weights loaded.")
except Exception as e:
    raise RuntimeError(
        "Could not load pretrained ImageNet weights. "
        "Check Kaggle Internet/settings or cache. "
        "Stopping rather than silently training from scratch."
    ) from e

# Freeze early layers; fine-tune layer3, layer4 and classifier.
for param in model.parameters():
    param.requires_grad = False

for param in model.layer3.parameters():
    param.requires_grad = True
for param in model.layer4.parameters():
    param.requires_grad = True

in_features = model.fc.in_features
model.fc = nn.Sequential(
    nn.Dropout(0.40),
    nn.Linear(in_features, len(CLASSES))
)

for param in model.fc.parameters():
    param.requires_grad = True

model = model.to(DEVICE)

# Moderate class weights; no WeightedRandomSampler in V3.
train_labels = np.array([
    CLASS_TO_IDX[label] for label in train_df["damage_label"]
])
counts = np.bincount(train_labels, minlength=len(CLASSES))

# Square-root inverse-frequency weighting, normalized to mean 1.
weights = 1.0 / np.sqrt(np.maximum(counts, 1))
weights = weights / weights.mean()

criterion = nn.CrossEntropyLoss(
    weight=torch.tensor(weights, dtype=torch.float32, device=DEVICE)
)

optimizer = torch.optim.AdamW([
    {"params": model.layer3.parameters(), "lr": BACKBONE_LR},
    {"params": model.layer4.parameters(), "lr": BACKBONE_LR},
    {"params": model.fc.parameters(), "lr": HEAD_LR}
], weight_decay=2e-4)

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode="max", factor=0.5, patience=1
)

# -------------------- 5. TRAIN / EVALUATION FUNCTION --------------------

def run_epoch(loader, training=False):
    if training:
        model.train()
    else:
        model.eval()

    total_loss = 0.0
    total = 0
    y_true, y_pred = [], []

    with torch.set_grad_enabled(training):
        for images, labels in loader:
            images = images.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)

            if training:
                optimizer.zero_grad(set_to_none=True)

            logits = model(images)
            loss = criterion(logits, labels)

            if training:
                loss.backward()
                torch.nn.utils.clip_grad_norm_(
                    [p for p in model.parameters() if p.requires_grad],
                    max_norm=1.0
                )
                optimizer.step()

            n = labels.size(0)
            total_loss += loss.item() * n
            total += n

            y_true.extend(labels.detach().cpu().numpy().tolist())
            y_pred.extend(logits.argmax(1).detach().cpu().numpy().tolist())

    return {
        "loss": total_loss / max(total, 1),
        "accuracy": accuracy_score(y_true, y_pred),
        "macro_f1": f1_score(
            y_true, y_pred, average="macro", zero_division=0
        ),
        "y_true": y_true,
        "y_pred": y_pred
    }

# -------------------- 6. TRAINING --------------------

history = []
best_f1 = -1.0
best_epoch = 0
best_state = None
bad_epochs = 0
start_time = time.time()

print("\n" + "=" * 65)
print("STARTING GEOVISION V3 TRAINING")
print("=" * 65)

for epoch in range(1, EPOCHS + 1):
    train_result = run_epoch(train_loader, training=True)
    val_result = run_epoch(val_loader, training=False)

    scheduler.step(val_result["macro_f1"])

    row = {
        "epoch": epoch,
        "train_loss": train_result["loss"],
        "train_accuracy": train_result["accuracy"],
        "train_macro_f1": train_result["macro_f1"],
        "val_loss": val_result["loss"],
        "val_accuracy": val_result["accuracy"],
        "val_macro_f1": val_result["macro_f1"]
    }
    history.append(row)

    print(
        f"Epoch {epoch:02d}/{EPOCHS} | "
        f"Train loss={row['train_loss']:.4f} "
        f"acc={row['train_accuracy']:.4f} | "
        f"Val loss={row['val_loss']:.4f} "
        f"acc={row['val_accuracy']:.4f} "
        f"macro-F1={row['val_macro_f1']:.4f}"
    )

    if val_result["macro_f1"] > best_f1:
        best_f1 = val_result["macro_f1"]
        best_epoch = epoch
        best_state = copy.deepcopy(model.state_dict())
        torch.save({
            "epoch": best_epoch,
            "model_state_dict": best_state,
            "classes": CLASSES,
            "image_size": IMAGE_SIZE,
            "validation_macro_f1": best_f1
        }, OUT / "geovision_resnet18_v3_best.pth")
        bad_epochs = 0
    else:
        bad_epochs += 1

    if bad_epochs >= PATIENCE:
        print("Early stopping.")
        break

pd.DataFrame(history).to_csv(OUT / "training_history.csv", index=False)

if best_state is None:
    raise RuntimeError("No valid best model checkpoint was saved.")

model.load_state_dict(best_state)
print(f"\nBest epoch: {best_epoch}")
print(f"Best validation macro-F1: {best_f1:.4f}")

# -------------------- 7. VALIDATION METRICS --------------------

val_result = run_epoch(val_loader, training=False)
val_report_text = classification_report(
    val_result["y_true"], val_result["y_pred"],
    labels=list(range(len(CLASSES))), target_names=CLASSES,
    zero_division=0
)
val_report_dict = classification_report(
    val_result["y_true"], val_result["y_pred"],
    labels=list(range(len(CLASSES))), target_names=CLASSES,
    output_dict=True, zero_division=0
)

pd.DataFrame(val_report_dict).transpose().to_csv(
    OUT / "validation_classification_report.csv"
)
pd.DataFrame(
    confusion_matrix(
        val_result["y_true"], val_result["y_pred"],
        labels=list(range(len(CLASSES)))
    ),
    index=CLASSES, columns=CLASSES
).to_csv(OUT / "validation_confusion_matrix.csv")

# -------------------- 8. HELD-OUT TEST EVALUATION --------------------
# Test is evaluated only after checkpoint selection.
test_result = run_epoch(test_loader, training=False)
test_report_text = classification_report(
    test_result["y_true"], test_result["y_pred"],
    labels=list(range(len(CLASSES))), target_names=CLASSES,
    zero_division=0
)
test_report_dict = classification_report(
    test_result["y_true"], test_result["y_pred"],
    labels=list(range(len(CLASSES))), target_names=CLASSES,
    output_dict=True, zero_division=0
)
test_cm = confusion_matrix(
    test_result["y_true"], test_result["y_pred"],
    labels=list(range(len(CLASSES)))
)

pd.DataFrame(test_report_dict).transpose().to_csv(
    OUT / "test_classification_report.csv"
)
pd.DataFrame(test_cm, index=CLASSES, columns=CLASSES).to_csv(
    OUT / "test_confusion_matrix.csv"
)

# -------------------- 9. PLOTS --------------------

hist = pd.DataFrame(history)

plt.figure(figsize=(9, 5))
plt.plot(hist["epoch"], hist["train_loss"], marker="o", label="Train")
plt.plot(hist["epoch"], hist["val_loss"], marker="o", label="Validation")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("GeoVision V3 — Loss")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(OUT / "loss_curve.png", dpi=160)
plt.close()

plt.figure(figsize=(9, 5))
plt.plot(hist["epoch"], hist["train_macro_f1"], marker="o", label="Train")
plt.plot(hist["epoch"], hist["val_macro_f1"], marker="o", label="Validation")
plt.xlabel("Epoch")
plt.ylabel("Macro-F1")
plt.title("GeoVision V3 — Macro-F1")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(OUT / "macro_f1_curve.png", dpi=160)
plt.close()

plt.figure(figsize=(7, 6))
plt.imshow(test_cm, interpolation="nearest")
plt.title("GeoVision V3 — Test Confusion Matrix")
plt.colorbar()
ticks = np.arange(len(CLASSES))
plt.xticks(ticks, CLASSES, rotation=35, ha="right")
plt.yticks(ticks, CLASSES)
plt.xlabel("Predicted")
plt.ylabel("Actual")
threshold = test_cm.max() / 2 if test_cm.max() else 0
for i in range(test_cm.shape[0]):
    for j in range(test_cm.shape[1]):
        plt.text(
            j, i, str(test_cm[i, j]),
            ha="center", va="center",
            color="white" if test_cm[i, j] > threshold else "black"
        )
plt.tight_layout()
plt.savefig(OUT / "test_confusion_matrix.png", dpi=160)
plt.close()

# -------------------- 10. SUMMARY + ZIP --------------------

elapsed = (time.time() - start_time) / 60
summary = {
    "project": "GeoVision",
    "version": "V3",
    "model": "ResNet18",
    "device": str(DEVICE),
    "classes": CLASSES,
    "image_size": IMAGE_SIZE,
    "best_epoch": int(best_epoch),
    "best_validation_macro_f1": float(best_f1),
    "validation_accuracy": float(val_result["accuracy"]),
    "validation_macro_f1": float(val_result["macro_f1"]),
    "test_accuracy": float(test_result["accuracy"]),
    "test_macro_f1": float(f1_score(
        test_result["y_true"], test_result["y_pred"],
        average="macro", zero_division=0
    )),
    "test_weighted_f1": float(f1_score(
        test_result["y_true"], test_result["y_pred"],
        average="weighted", zero_division=0
    )),
    "train_samples": len(train_df),
    "validation_samples": len(val_df),
    "test_samples": len(test_df),
    "train_disasters": sorted(train_dis),
    "validation_disasters": sorted(val_dis),
    "test_disasters": sorted(test_dis),
    "training_minutes": round(elapsed, 2),
    "note": (
        "Uses the same splits as V2. Test results are for development "
        "comparison and are not an untouched final benchmark."
    )
}

with open(OUT / "run_summary.json", "w") as f:
    json.dump(summary, f, indent=2)

with open(OUT / "evaluation_metrics.json", "w") as f:
    json.dump({
        **summary,
        "validation_classification_report": val_report_dict,
        "test_classification_report": test_report_dict
    }, f, indent=2)

with open(OUT / "environment.txt", "w") as f:
    f.write(f"PyTorch: {torch.__version__}\n")
    f.write(f"Device: {DEVICE}\n")
    if DEVICE.type == "cuda":
        f.write(f"GPU: {torch.cuda.get_device_name(0)}\n")

zip_path = Path("/kaggle/working/GeoVision_V3_Results.zip")
if zip_path.exists():
    zip_path.unlink()

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as z:
    for file in OUT.rglob("*"):
        if file.is_file():
            z.write(
                file,
                arcname=f"GeoVision_V3_Results/{file.relative_to(OUT)}"
            )

# -------------------- 11. FINAL REPORT --------------------

print("\n" + "=" * 65)
print("GEOVISION V3 — FINAL RESULTS")
print("=" * 65)
print("Best epoch:", best_epoch)
print(f"Validation accuracy: {summary['validation_accuracy']:.4f}")
print(f"Validation macro-F1: {summary['validation_macro_f1']:.4f}")
print(f"Test accuracy:       {summary['test_accuracy']:.4f}")
print(f"Test macro-F1:       {summary['test_macro_f1']:.4f}")
print(f"Test weighted-F1:    {summary['test_weighted_f1']:.4f}")
print(f"Training time:       {elapsed:.2f} minutes")

print("\nVALIDATION CLASSIFICATION REPORT")
print(val_report_text)
print("\nTEST CLASSIFICATION REPORT")
print(test_report_text)

print("\nResults folder:", OUT)
print("Results ZIP:", zip_path)
print("ZIP size:", round(zip_path.stat().st_size / (1024 * 1024), 2), "MB")
print("\nGeoVision V3 training completed.")
```
